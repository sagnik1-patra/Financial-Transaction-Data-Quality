```python
import os
import json
import random
import warnings
from pathlib import Path

import numpy as np
import pandas as pd

import matplotlib.pyplot as plt

import joblib
import yaml

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    confusion_matrix,
    classification_report
)

import tensorflow as tf

from tensorflow.keras.models import Sequential

from tensorflow.keras.layers import (
    Dense,
    Dropout,
    BatchNormalization
)

from tensorflow.keras.callbacks import (
    EarlyStopping,
    ReduceLROnPlateau
)


# =====================================================================
# 2. GLOBAL SETTINGS
# =====================================================================

warnings.filterwarnings("ignore")

SEED = 42

random.seed(SEED)
np.random.seed(SEED)
tf.random.set_seed(SEED)


# =====================================================================
# 3. PROJECT PATHS
# =====================================================================

PROJECT_DIR = Path(
    r"C:\Users\sagni\Downloads\Financial Transaction Data Quality"
)

CSV_PATH = (
    PROJECT_DIR
    / "dirty_financial_transactions.csv"
)


# =====================================================================
# 4. OUTPUT PATHS
# =====================================================================

H5_PATH = (
    PROJECT_DIR
    / "financial_data_quality_model.h5"
)

PKL_PATH = (
    PROJECT_DIR
    / "financial_data_quality_pipeline.pkl"
)

YAML_PATH = (
    PROJECT_DIR
    / "financial_data_quality_config.yaml"
)

JSON_PATH = (
    PROJECT_DIR
    / "financial_data_quality_results.json"
)


# CSV OUTPUTS

RESULT_CSV_PATH = (
    PROJECT_DIR
    / "result.csv"
)

PREDICTION_CSV_PATH = (
    PROJECT_DIR
    / "prediction.csv"
)

LABELED_CSV_PATH = (
    PROJECT_DIR
    / "financial_data_quality_labeled.csv"
)


# GRAPH OUTPUTS

ACCURACY_GRAPH_PATH = (
    PROJECT_DIR
    / "accuracy_graph.png"
)

HEATMAP_PATH = (
    PROJECT_DIR
    / "heatmap.png"
)

COMPARISON_GRAPH_PATH = (
    PROJECT_DIR
    / "comparison_graph.png"
)

RESULT_GRAPH_PATH = (
    PROJECT_DIR
    / "result_graph.png"
)

PREDICTION_GRAPH_PATH = (
    PROJECT_DIR
    / "prediction_graph.png"
)


PROJECT_DIR.mkdir(
    parents=True,
    exist_ok=True
)


# =====================================================================
# 5. LOAD DATA
# =====================================================================

print("\n" + "=" * 80)
print("AI-BASED FINANCIAL TRANSACTION DATA QUALITY PREDICTION")
print("=" * 80)


if not CSV_PATH.exists():

    raise FileNotFoundError(
        f"Dataset not found:\n{CSV_PATH}"
    )


print("\nLoading dataset...")


df = pd.read_csv(
    CSV_PATH,
    low_memory=False
)


original_df = df.copy()


print("\nDataset loaded successfully.")

print(
    "Dataset shape:",
    df.shape
)


print("\nColumns:")

for col in df.columns:

    print(
        " -",
        col
    )


# =====================================================================
# 6. VALIDATE COLUMNS
# =====================================================================

required_columns = [

    "Transaction_ID",
    "Transaction_Date",
    "Customer_ID",
    "Product_Name",
    "Quantity",
    "Price",
    "Payment_Method",
    "Transaction_Status"

]


missing_columns = [

    col
    for col in required_columns
    if col not in df.columns

]


if missing_columns:

    raise ValueError(
        f"Missing columns: {missing_columns}"
    )


# =====================================================================
# 7. HELPER FUNCTIONS
# =====================================================================


def safe_string(value):

    if pd.isna(value):

        return ""

    return str(value).strip()



def normalize_text(value):

    return safe_string(
        value
    ).lower().strip()



def clean_price(value):

    if pd.isna(value):

        return np.nan

    value = str(value)

    value = value.replace(
        "$",
        ""
    )

    value = value.replace(
        ",",
        ""
    )

    value = value.strip()

    try:

        return float(value)

    except:

        return np.nan



def valid_transaction_id(value):

    value = safe_string(
        value
    )

    if value == "":

        return False

    if not value.startswith("T"):

        return False

    numeric_part = value[1:]

    return numeric_part.isdigit()



def valid_customer_id(value):

    value = safe_string(
        value
    )

    if value == "":

        return False

    return True



# =====================================================================
# 8. VALID STANDARD VALUES
# =====================================================================

VALID_PRODUCTS = {

    "Laptop",
    "Tablet",
    "Smartphone",
    "Headphones",
    "Coffee Machine"

}


VALID_PAYMENT_METHODS = {

    "Credit Card",
    "PayPal",
    "Cash"

}


VALID_STATUSES = {

    "Completed",
    "Pending",
    "Failed"

}


# =====================================================================
# 9. DATA PREPROCESSING
# =====================================================================

df["Price_Numeric"] = (

    df["Price"]
    .apply(clean_price)

)


df["Quantity_Numeric"] = pd.to_numeric(

    df["Quantity"],

    errors="coerce"

)


df["Parsed_Date"] = pd.to_datetime(

    df["Transaction_Date"],

    errors="coerce"

)


# =====================================================================
# 10. ORIGINAL CATEGORY VALUES
# =====================================================================

product_original = (

    df["Product_Name"]
    .fillna("")
    .astype(str)
    .str.strip()

)


payment_original = (

    df["Payment_Method"]
    .fillna("")
    .astype(str)
    .str.strip()

)


status_original = (

    df["Transaction_Status"]
    .fillna("")
    .astype(str)
    .str.strip()

)


# =====================================================================
# 11. CREATE ERROR / WRONGNESS RULES
# =====================================================================


# -----------------------------
# Transaction ID
# -----------------------------

df["Error_Missing_Transaction_ID"] = (

    df["Transaction_ID"]
    .isna()
    .astype(int)

)


df["Error_Invalid_Transaction_ID"] = (

    ~df["Transaction_ID"]
    .apply(valid_transaction_id)

).astype(int)


df["Error_Duplicate_Transaction_ID"] = (

    (
        df["Transaction_ID"]
        .duplicated(
            keep=False
        )
    )

    &

    (
        df["Transaction_ID"]
        .notna()
    )

).astype(int)


# -----------------------------
# Customer ID
# -----------------------------

df["Error_Missing_Customer_ID"] = (

    df["Customer_ID"]
    .isna()
    .astype(int)

)


df["Error_Invalid_Customer_ID"] = (

    ~df["Customer_ID"]
    .apply(valid_customer_id)

).astype(int)


# -----------------------------
# Date
# -----------------------------

df["Error_Missing_Date"] = (

    df["Transaction_Date"]
    .isna()
    .astype(int)

)


df["Error_Invalid_Date"] = (

    (
        df["Transaction_Date"]
        .notna()
    )

    &

    (
        df["Parsed_Date"]
        .isna()
    )

).astype(int)


# -----------------------------
# Product
# -----------------------------

df["Error_Missing_Product"] = (

    df["Product_Name"]
    .isna()
    .astype(int)

)


df["Error_Invalid_Product"] = (

    (
        df["Product_Name"]
        .notna()
    )

    &

    (
        ~product_original.isin(
            VALID_PRODUCTS
        )
    )

).astype(int)


# -----------------------------
# Quantity
# -----------------------------

df["Error_Missing_Quantity"] = (

    df["Quantity_Numeric"]
    .isna()
    .astype(int)

)


df["Error_Invalid_Quantity"] = (

    (
        df["Quantity_Numeric"]
        .notna()
    )

    &

    (
        df["Quantity_Numeric"]
        <= 0
    )

).astype(int)


# -----------------------------
# Price
# -----------------------------

df["Error_Missing_Price"] = (

    df["Price"]
    .isna()
    .astype(int)

)


df["Error_Invalid_Price_Format"] = (

    (
        df["Price"]
        .notna()
    )

    &

    (
        df["Price_Numeric"]
        .isna()
    )

).astype(int)


df["Error_Invalid_Price_Value"] = (

    (
        df["Price_Numeric"]
        .notna()
    )

    &

    (
        df["Price_Numeric"]
        <= 0
    )

).astype(int)


# -----------------------------
# Payment
# -----------------------------

df["Error_Missing_Payment_Method"] = (

    df["Payment_Method"]
    .isna()
    .astype(int)

)


df["Error_Nonstandard_Payment"] = (

    (
        df["Payment_Method"]
        .notna()
    )

    &

    (
        ~payment_original.isin(
            VALID_PAYMENT_METHODS
        )
    )

).astype(int)


# -----------------------------
# Status
# -----------------------------

df["Error_Missing_Status"] = (

    df["Transaction_Status"]
    .isna()
    .astype(int)

)


df["Error_Nonstandard_Status"] = (

    (
        df["Transaction_Status"]
        .notna()
    )

    &

    (
        ~status_original.isin(
            VALID_STATUSES
        )
    )

).astype(int)


# =====================================================================
# 12. ERROR COLUMN LIST
# =====================================================================

ERROR_COLUMNS = [

    "Error_Missing_Transaction_ID",

    "Error_Invalid_Transaction_ID",

    "Error_Duplicate_Transaction_ID",

    "Error_Missing_Customer_ID",

    "Error_Invalid_Customer_ID",

    "Error_Missing_Date",

    "Error_Invalid_Date",

    "Error_Missing_Product",

    "Error_Invalid_Product",

    "Error_Missing_Quantity",

    "Error_Invalid_Quantity",

    "Error_Missing_Price",

    "Error_Invalid_Price_Format",

    "Error_Invalid_Price_Value",

    "Error_Missing_Payment_Method",

    "Error_Nonstandard_Payment",

    "Error_Missing_Status",

    "Error_Nonstandard_Status"

]


ERROR_NAMES = {

    "Error_Missing_Transaction_ID":
        "Missing Transaction ID",

    "Error_Invalid_Transaction_ID":
        "Invalid Transaction ID",

    "Error_Duplicate_Transaction_ID":
        "Duplicate Transaction ID",

    "Error_Missing_Customer_ID":
        "Missing Customer ID",

    "Error_Invalid_Customer_ID":
        "Invalid Customer ID",

    "Error_Missing_Date":
        "Missing Date",

    "Error_Invalid_Date":
        "Invalid Date",

    "Error_Missing_Product":
        "Missing Product",

    "Error_Invalid_Product":
        "Invalid Product",

    "Error_Missing_Quantity":
        "Missing Quantity",

    "Error_Invalid_Quantity":
        "Invalid Quantity",

    "Error_Missing_Price":
        "Missing Price",

    "Error_Invalid_Price_Format":
        "Invalid Price Format",

    "Error_Invalid_Price_Value":
        "Invalid Price Value",

    "Error_Missing_Payment_Method":
        "Missing Payment Method",

    "Error_Nonstandard_Payment":
        "Nonstandard Payment Method",

    "Error_Missing_Status":
        "Missing Status",

    "Error_Nonstandard_Status":
        "Nonstandard Status"

}


# =====================================================================
# 13. CREATE TARGET
# =====================================================================

df["Error_Count"] = (

    df[ERROR_COLUMNS]
    .sum(
        axis=1
    )

)


df["Is_Anomaly"] = (

    df["Error_Count"] > 0

).astype(int)


# =====================================================================
# 14. ERROR DESCRIPTION
# =====================================================================


def create_error_description(row):

    found_errors = []

    for error_column in ERROR_COLUMNS:

        if row[error_column] == 1:

            found_errors.append(

                ERROR_NAMES[
                    error_column
                ]

            )

    if len(found_errors) == 0:

        return "Clean"

    return " | ".join(
        found_errors
    )


df["Error_Type"] = df.apply(

    create_error_description,

    axis=1

)


# =====================================================================
# 15. DATA SUMMARY
# =====================================================================

total_records = len(df)


clean_records = int(

    (
        df["Is_Anomaly"] == 0
    ).sum()

)


anomaly_records = int(

    (
        df["Is_Anomaly"] == 1
    ).sum()

)


print("\n" + "=" * 80)
print("DATA QUALITY SUMMARY")
print("=" * 80)


print(
    "\nTotal records:",
    total_records
)

print(
    "Clean records:",
    clean_records
)

print(
    "Wrong / anomalous records:",
    anomaly_records
)


print(
    "Clean percentage:",
    round(
        clean_records
        / total_records
        * 100,
        2
    ),
    "%"
)


print(
    "Anomaly percentage:",
    round(
        anomaly_records
        / total_records
        * 100,
        2
    ),
    "%"
)


# =====================================================================
# 16. FEATURE ENGINEERING
# =====================================================================

features = pd.DataFrame(
    index=df.index
)


# Transaction ID

features[
    "transaction_id_missing"
] = (

    df["Transaction_ID"]
    .isna()
    .astype(int)

)


features[
    "transaction_id_length"
] = (

    df["Transaction_ID"]
    .fillna("")
    .astype(str)
    .str.len()

)


features[
    "transaction_id_valid"
] = (

    df["Transaction_ID"]
    .apply(valid_transaction_id)
    .astype(int)

)


features[
    "transaction_id_duplicate"
] = (

    df["Transaction_ID"]
    .duplicated(
        keep=False
    )
    .astype(int)

)


# Customer ID

features[
    "customer_missing"
] = (

    df["Customer_ID"]
    .isna()
    .astype(int)

)


features[
    "customer_id_length"
] = (

    df["Customer_ID"]
    .fillna("")
    .astype(str)
    .str.len()

)


# Date

features[
    "date_missing"
] = (

    df["Transaction_Date"]
    .isna()
    .astype(int)

)


features[
    "date_valid"
] = (

    df["Parsed_Date"]
    .notna()
    .astype(int)

)


features[
    "year"
] = (

    df["Parsed_Date"]
    .dt.year
    .fillna(0)

)


features[
    "month"
] = (

    df["Parsed_Date"]
    .dt.month
    .fillna(0)

)


features[
    "day"
] = (

    df["Parsed_Date"]
    .dt.day
    .fillna(0)

)


# Product

features[
    "product_missing"
] = (

    df["Product_Name"]
    .isna()
    .astype(int)

)


features[
    "product_length"
] = (

    product_original
    .str.len()

)


features[
    "product_valid"
] = (

    product_original
    .isin(
        VALID_PRODUCTS
    )
    .astype(int)

)


# Quantity

features[
    "quantity_missing"
] = (

    df["Quantity_Numeric"]
    .isna()
    .astype(int)

)


features[
    "quantity_value"
] = (

    df["Quantity_Numeric"]
    .fillna(0)

)


features[
    "quantity_positive"
] = (

    df["Quantity_Numeric"]
    .gt(0)
    .fillna(False)
    .astype(int)

)


features[
    "quantity_negative"
] = (

    df["Quantity_Numeric"]
    .lt(0)
    .fillna(False)
    .astype(int)

)


# Price

features[
    "price_missing"
] = (

    df["Price"]
    .isna()
    .astype(int)

)


features[
    "price_value"
] = (

    df["Price_Numeric"]
    .fillna(0)

)


features[
    "price_positive"
] = (

    df["Price_Numeric"]
    .gt(0)
    .fillna(False)
    .astype(int)

)


features[
    "price_negative"
] = (

    df["Price_Numeric"]
    .lt(0)
    .fillna(False)
    .astype(int)

)


features[
    "price_has_dollar"
] = (

    df["Price"]
    .fillna("")
    .astype(str)
    .str.contains(
        r"\$",
        regex=True
    )
    .astype(int)

)


features[
    "price_length"
] = (

    df["Price"]
    .fillna("")
    .astype(str)
    .str.len()

)


# Payment

features[
    "payment_missing"
] = (

    df["Payment_Method"]
    .isna()
    .astype(int)

)


features[
    "payment_valid"
] = (

    payment_original
    .isin(
        VALID_PAYMENT_METHODS
    )
    .astype(int)

)


features[
    "payment_length"
] = (

    payment_original
    .str.len()

)


# Status

features[
    "status_missing"
] = (

    df["Transaction_Status"]
    .isna()
    .astype(int)

)


features[
    "status_valid"
] = (

    status_original
    .isin(
        VALID_STATUSES
    )
    .astype(int)

)


features[
    "status_length"
] = (

    status_original
    .str.len()

)


# =====================================================================
# 17. CLEAN ML FEATURES
# =====================================================================

features = features.replace(

    [
        np.inf,
        -np.inf
    ],

    np.nan

)


features = features.fillna(
    0
)


features = features.astype(
    np.float32
)


X = features


y = (

    df["Is_Anomaly"]
    .astype(
        np.int32
    )

)


print(
    "\nNumber of ML features:",
    X.shape[1]
)


# =====================================================================
# 18. TRAIN TEST SPLIT
# =====================================================================

X_train, X_test, y_train, y_test = train_test_split(

    X,
    y,

    test_size=0.20,

    random_state=SEED,

    stratify=y

)


print(
    "\nTraining records:",
    len(X_train)
)

print(
    "Testing records:",
    len(X_test)
)


# =====================================================================
# 19. SCALE FEATURES
# =====================================================================

scaler = StandardScaler()


X_train_scaled = scaler.fit_transform(
    X_train
)


X_test_scaled = scaler.transform(
    X_test
)


# =====================================================================
# 20. CLASS WEIGHTS
# =====================================================================

class_counts = np.bincount(
    y_train
)


class_weight = None


if len(class_counts) == 2:

    total = len(y_train)

    class_weight = {

        0:

            total
            /
            (
                2
                *
                class_counts[0]
            ),

        1:

            total
            /
            (
                2
                *
                class_counts[1]
            )

    }


# =====================================================================
# 21. BUILD ANN MODEL
# =====================================================================

model = Sequential(

    [

        tf.keras.Input(

            shape=(
                X_train_scaled.shape[1],
            )

        ),

        Dense(
            128,
            activation="relu"
        ),

        BatchNormalization(),

        Dropout(
            0.30
        ),

        Dense(
            64,
            activation="relu"
        ),

        BatchNormalization(),

        Dropout(
            0.25
        ),

        Dense(
            32,
            activation="relu"
        ),

        Dropout(
            0.20
        ),

        Dense(
            16,
            activation="relu"
        ),

        Dense(
            1,
            activation="sigmoid"
        )

    ]

)


# =====================================================================
# 22. COMPILE MODEL
# =====================================================================

model.compile(

    optimizer=tf.keras.optimizers.Adam(

        learning_rate=0.001

    ),

    loss="binary_crossentropy",

    metrics=[

        "accuracy",

        tf.keras.metrics.Precision(
            name="precision"
        ),

        tf.keras.metrics.Recall(
            name="recall"
        ),

        tf.keras.metrics.AUC(
            name="auc"
        )

    ]

)


model.summary()


# =====================================================================
# 23. CALLBACKS
# =====================================================================

early_stopping = EarlyStopping(

    monitor="val_loss",

    patience=7,

    restore_best_weights=True,

    verbose=1

)


reduce_lr = ReduceLROnPlateau(

    monitor="val_loss",

    factor=0.5,

    patience=3,

    min_lr=1e-6,

    verbose=1

)


# =====================================================================
# 24. TRAIN MODEL
# =====================================================================

print("\n" + "=" * 80)
print("MODEL TRAINING")
print("=" * 80)


history = model.fit(

    X_train_scaled,
    y_train,

    validation_split=0.20,

    epochs=50,

    batch_size=256,

    callbacks=[

        early_stopping,
        reduce_lr

    ],

    class_weight=class_weight,

    verbose=1

)


# =====================================================================
# 25. MAKE PREDICTIONS
# =====================================================================

prediction_probability = (

    model.predict(

        X_test_scaled,

        verbose=0

    )

    .flatten()

)


prediction_label = (

    prediction_probability
    >= 0.50

).astype(int)


# =====================================================================
# 26. CALCULATE MODEL METRICS
# =====================================================================

accuracy = accuracy_score(

    y_test,
    prediction_label

)


precision = precision_score(

    y_test,
    prediction_label,

    zero_division=0

)


recall = recall_score(

    y_test,
    prediction_label,

    zero_division=0

)


f1 = f1_score(

    y_test,
    prediction_label,

    zero_division=0

)


try:

    roc_auc = roc_auc_score(

        y_test,
        prediction_probability

    )

except:

    roc_auc = 0


cm = confusion_matrix(

    y_test,
    prediction_label

)


print("\n" + "=" * 80)
print("FINAL MODEL RESULTS")
print("=" * 80)


print(
    f"\nAccuracy : {accuracy * 100:.2f}%"
)

print(
    f"Precision: {precision * 100:.2f}%"
)

print(
    f"Recall   : {recall * 100:.2f}%"
)

print(
    f"F1 Score : {f1 * 100:.2f}%"
)

print(
    f"ROC-AUC  : {roc_auc * 100:.2f}%"
)


print("\nConfusion Matrix:")

print(
    cm
)


print("\nClassification Report:")


print(

    classification_report(

        y_test,
        prediction_label,

        target_names=[

            "Clean",
            "Anomaly"

        ],

        zero_division=0

    )

)


# =====================================================================
# 27. ACCURACY GRAPH
# =====================================================================
#
# Training Accuracy vs Validation Accuracy
#
# Saves:
# accuracy_graph.png
#
# Shows graph also.
#
# =====================================================================

print(
    "\nGenerating accuracy graph..."
)


epochs_range = range(

    1,
    len(
        history.history["accuracy"]
    ) + 1

)


plt.figure(
    figsize=(10, 6)
)


plt.plot(

    epochs_range,

    history.history[
        "accuracy"
    ],

    marker="o",

    markersize=3,

    label="Training Accuracy"

)


plt.plot(

    epochs_range,

    history.history[
        "val_accuracy"
    ],

    marker="o",

    markersize=3,

    label="Validation Accuracy"

)


plt.xlabel(
    "Epoch"
)

plt.ylabel(
    "Accuracy"
)

plt.title(
    "Financial Data Quality Model - Accuracy Graph"
)

plt.legend()

plt.grid(
    True,
    alpha=0.3
)

plt.tight_layout()


plt.savefig(

    ACCURACY_GRAPH_PATH,

    dpi=300,

    bbox_inches="tight"

)


print(
    "Saved:",
    ACCURACY_GRAPH_PATH
)


plt.show()

plt.close()


# =====================================================================
# 28. CONFUSION MATRIX HEATMAP
# =====================================================================
#
# No seaborn required.
#
# Saves:
# heatmap.png
#
# =====================================================================

print(
    "\nGenerating confusion matrix heatmap..."
)


plt.figure(
    figsize=(7, 6)
)


plt.imshow(
    cm
)


plt.title(
    "Confusion Matrix Heatmap"
)


plt.xlabel(
    "Predicted Label"
)

plt.ylabel(
    "Actual Label"
)


plt.xticks(

    [0, 1],

    [
        "Clean",
        "Anomaly"
    ]

)


plt.yticks(

    [0, 1],

    [
        "Clean",
        "Anomaly"
    ]

)


for i in range(
    cm.shape[0]
):

    for j in range(
        cm.shape[1]
    ):

        plt.text(

            j,
            i,

            str(
                cm[i, j]
            ),

            ha="center",
            va="center",

            fontsize=14

        )


plt.colorbar(
    label="Number of Transactions"
)


plt.tight_layout()


plt.savefig(

    HEATMAP_PATH,

    dpi=300,

    bbox_inches="tight"

)


print(
    "Saved:",
    HEATMAP_PATH
)


plt.show()

plt.close()


# =====================================================================
# 29. COMPARISON GRAPH
# =====================================================================
#
# Accuracy vs Precision vs Recall vs F1 vs ROC-AUC
#
# Saves:
# comparison_graph.png
#
# =====================================================================

print(
    "\nGenerating comparison graph..."
)


metric_names = [

    "Accuracy",
    "Precision",
    "Recall",
    "F1 Score",
    "ROC-AUC"

]


metric_values = [

    accuracy * 100,
    precision * 100,
    recall * 100,
    f1 * 100,
    roc_auc * 100

]


plt.figure(
    figsize=(10, 6)
)


bars = plt.bar(

    metric_names,

    metric_values

)


plt.xlabel(
    "Evaluation Metric"
)

plt.ylabel(
    "Score (%)"
)

plt.title(
    "Model Performance Comparison"
)


plt.ylim(
    0,
    105
)


for bar, value in zip(
    bars,
    metric_values
):

    plt.text(

        bar.get_x()
        +
        bar.get_width() / 2,

        value + 1,

        f"{value:.2f}%",

        ha="center",

        fontsize=10

    )


plt.grid(

    axis="y",

    alpha=0.3

)


plt.tight_layout()


plt.savefig(

    COMPARISON_GRAPH_PATH,

    dpi=300,

    bbox_inches="tight"

)


print(
    "Saved:",
    COMPARISON_GRAPH_PATH
)


plt.show()

plt.close()


# =====================================================================
# 30. CREATE RESULT CSV
# =====================================================================
#
# result.csv contains data-quality error category summary.
#
# =====================================================================

print(
    "\nCreating result.csv..."
)


result_records = []


for column in ERROR_COLUMNS:

    count = int(

        df[column]
        .sum()

    )


    percentage = (

        count
        /
        total_records
        *
        100

    )


    result_records.append(

        {

            "Error_Type":
                ERROR_NAMES[column],

            "Error_Count":
                count,

            "Percentage":
                round(
                    percentage,
                    4
                )

        }

    )


result_df = pd.DataFrame(
    result_records
)


result_df = result_df.sort_values(

    by="Error_Count",

    ascending=False

).reset_index(
    drop=True
)


result_df.to_csv(

    RESULT_CSV_PATH,

    index=False

)


print(
    "Saved:",
    RESULT_CSV_PATH
)


print(
    "\nResult CSV:"
)


print(
    result_df
)


# =====================================================================
# 31. RESULT GRAPH
# =====================================================================
#
# Graph of errors found in transaction dataset.
#
# Saves:
# result_graph.png
#
# =====================================================================

print(
    "\nGenerating result graph..."
)


plt.figure(
    figsize=(13, 8)
)


bars = plt.barh(

    result_df[
        "Error_Type"
    ],

    result_df[
        "Error_Count"
    ]

)


plt.xlabel(
    "Number of Records"
)

plt.ylabel(
    "Data Quality Error"
)

plt.title(
    "Financial Transaction Data Quality Error Distribution"
)


plt.gca().invert_yaxis()


for bar in bars:

    width = bar.get_width()

    plt.text(

        width,

        bar.get_y()
        +
        bar.get_height() / 2,

        f" {int(width)}",

        va="center"

    )


plt.grid(

    axis="x",

    alpha=0.3

)


plt.tight_layout()


plt.savefig(

    RESULT_GRAPH_PATH,

    dpi=300,

    bbox_inches="tight"

)


print(
    "Saved:",
    RESULT_GRAPH_PATH
)


plt.show()

plt.close()


# =====================================================================
# 32. CREATE PREDICTION CSV
# =====================================================================

print(
    "\nCreating prediction.csv..."
)


prediction_df = (

    original_df
    .loc[
        X_test.index
    ]
    .copy()

)


prediction_df[
    "Actual_Label"
] = y_test.values


prediction_df[
    "Actual_Result"
] = np.where(

    y_test.values == 1,

    "Anomaly",

    "Clean"

)


prediction_df[
    "Prediction_Probability"
] = prediction_probability


prediction_df[
    "Predicted_Label"
] = prediction_label


prediction_df[
    "Predicted_Result"
] = np.where(

    prediction_label == 1,

    "Anomaly",

    "Clean"

)


prediction_df[
    "Correct_Prediction"
] = (

    prediction_df[
        "Actual_Label"
    ]

    ==

    prediction_df[
        "Predicted_Label"
    ]

)


# Add original detected error information

prediction_df[
    "Error_Count"
] = df.loc[
    X_test.index,
    "Error_Count"
].values


prediction_df[
    "Error_Type"
] = df.loc[
    X_test.index,
    "Error_Type"
].values


prediction_df.to_csv(

    PREDICTION_CSV_PATH,

    index=False

)


print(
    "Saved:",
    PREDICTION_CSV_PATH
)


print(
    "\nFirst 10 predictions:"
)


print(

    prediction_df[
        [

            "Transaction_ID",

            "Actual_Result",

            "Prediction_Probability",

            "Predicted_Result",

            "Correct_Prediction"

        ]
    ].head(
        10
    )

)


# =====================================================================
# 33. PREDICTION GRAPH
# =====================================================================
#
# Show first 100 test predictions.
#
# 0 = Clean
# 1 = Anomaly
#
# =====================================================================

print(
    "\nGenerating prediction graph..."
)


number_to_plot = min(

    100,

    len(
        prediction_df
    )

)


plot_predictions = prediction_df.iloc[
    :number_to_plot
]


x_positions = np.arange(
    number_to_plot
)


plt.figure(
    figsize=(15, 7)
)


plt.plot(

    x_positions,

    plot_predictions[
        "Actual_Label"
    ],

    marker="o",

    markersize=5,

    linewidth=1.5,

    label="Actual Label"

)


plt.plot(

    x_positions,

    plot_predictions[
        "Prediction_Probability"
    ],

    marker=".",

    markersize=4,

    linewidth=1.5,

    label="Predicted Probability"

)


plt.axhline(

    y=0.5,

    linestyle="--",

    label="Prediction Threshold = 0.5"

)


plt.xlabel(
    "Test Transaction Sample"
)

plt.ylabel(
    "Prediction Value"
)


plt.title(
    "Actual vs Predicted Financial Transaction Wrongness"
)


plt.yticks(

    [
        0,
        0.5,
        1
    ],

    [
        "Clean (0)",
        "Threshold",
        "Anomaly (1)"
    ]

)


plt.legend()


plt.grid(
    True,
    alpha=0.3
)


plt.tight_layout()


plt.savefig(

    PREDICTION_GRAPH_PATH,

    dpi=300,

    bbox_inches="tight"

)


print(
    "Saved:",
    PREDICTION_GRAPH_PATH
)


plt.show()

plt.close()


# =====================================================================
# 34. SAVE FULL LABELED DATASET
# =====================================================================

labeled_df = original_df.copy()


for column in ERROR_COLUMNS:

    labeled_df[
        column
    ] = df[
        column
    ]


labeled_df[
    "Error_Count"
] = df[
    "Error_Count"
]


labeled_df[
    "Error_Type"
] = df[
    "Error_Type"
]


labeled_df[
    "Is_Anomaly"
] = df[
    "Is_Anomaly"
]


labeled_df.to_csv(

    LABELED_CSV_PATH,

    index=False

)


print(
    "\nSaved labeled dataset:"
)

print(
    LABELED_CSV_PATH
)


# =====================================================================
# 35. SAVE H5 MODEL
# =====================================================================

model.save(
    H5_PATH
)


print(
    "\nSaved H5:"
)

print(
    H5_PATH
)


# =====================================================================
# 36. SAVE PKL PIPELINE
# =====================================================================

pipeline = {

    "scaler":
        scaler,

    "feature_names":
        list(
            X.columns
        ),

    "prediction_threshold":
        0.50,

    "valid_products":
        list(
            VALID_PRODUCTS
        ),

    "valid_payment_methods":
        list(
            VALID_PAYMENT_METHODS
        ),

    "valid_statuses":
        list(
            VALID_STATUSES
        ),

    "error_columns":
        ERROR_COLUMNS,

    "error_names":
        ERROR_NAMES

}


joblib.dump(

    pipeline,

    PKL_PATH

)


print(
    "\nSaved PKL:"
)

print(
    PKL_PATH
)


# =====================================================================
# 37. SAVE YAML CONFIG
# =====================================================================

config = {

    "project": {

        "name":
            "AI-Based Financial Transaction Data Quality Prediction",

        "dataset":
            str(
                CSV_PATH
            ),

        "output_directory":
            str(
                PROJECT_DIR
            )

    },


    "dataset": {

        "total_records":
            int(
                total_records
            ),

        "total_features":
            int(
                original_df.shape[1]
            ),

        "clean_records":
            int(
                clean_records
            ),

        "anomaly_records":
            int(
                anomaly_records
            )

    },


    "model": {

        "algorithm":
            "Artificial Neural Network",

        "framework":
            "TensorFlow Keras",

        "input_features":
            int(
                X.shape[1]
            ),

        "hidden_layers": [

            128,
            64,
            32,
            16

        ],

        "output_activation":
            "sigmoid",

        "optimizer":
            "Adam",

        "loss":
            "binary_crossentropy",

        "maximum_epochs":
            50,

        "batch_size":
            256,

        "prediction_threshold":
            0.5

    },


    "results": {

        "accuracy":
            float(
                accuracy
            ),

        "precision":
            float(
                precision
            ),

        "recall":
            float(
                recall
            ),

        "f1_score":
            float(
                f1
            ),

        "roc_auc":
            float(
                roc_auc
            )

    },


    "output_files": {

        "model":
            H5_PATH.name,

        "pipeline":
            PKL_PATH.name,

        "configuration":
            YAML_PATH.name,

        "json_results":
            JSON_PATH.name,

        "accuracy_graph":
            ACCURACY_GRAPH_PATH.name,

        "heatmap":
            HEATMAP_PATH.name,

        "comparison_graph":
            COMPARISON_GRAPH_PATH.name,

        "result_csv":
            RESULT_CSV_PATH.name,

        "result_graph":
            RESULT_GRAPH_PATH.name,

        "prediction_csv":
            PREDICTION_CSV_PATH.name,

        "prediction_graph":
            PREDICTION_GRAPH_PATH.name

    }

}


with open(

    YAML_PATH,

    "w",

    encoding="utf-8"

) as file:

    yaml.safe_dump(

        config,

        file,

        sort_keys=False,

        allow_unicode=True

    )


print(
    "\nSaved YAML:"
)

print(
    YAML_PATH
)


# =====================================================================
# 38. SAVE JSON RESULTS
# =====================================================================

history_json = {}


for key, values in history.history.items():

    history_json[
        key
    ] = [

        float(
            x
        )

        for x in values

    ]


json_data = {

    "project":
        "AI-Based Financial Transaction Data Quality Prediction",

    "total_records":
        int(
            total_records
        ),

    "clean_records":
        int(
            clean_records
        ),

    "anomaly_records":
        int(
            anomaly_records
        ),

    "training_records":
        int(
            len(
                X_train
            )
        ),

    "testing_records":
        int(
            len(
                X_test
            )
        ),

    "metrics": {

        "accuracy":
            float(
                accuracy
            ),

        "precision":
            float(
                precision
            ),

        "recall":
            float(
                recall
            ),

        "f1_score":
            float(
                f1
            ),

        "roc_auc":
            float(
                roc_auc
            )

    },

    "confusion_matrix":
        cm.tolist(),

    "training_history":
        history_json,

    "error_distribution":
        {

            ERROR_NAMES[column]:
                int(
                    df[column]
                    .sum()
                )

            for column
            in ERROR_COLUMNS

        }

}


with open(

    JSON_PATH,

    "w",

    encoding="utf-8"

) as file:

    json.dump(

        json_data,

        file,

        indent=4,

        ensure_ascii=False

    )


print(
    "\nSaved JSON:"
)

print(
    JSON_PATH
)


# =====================================================================
# 39. VERIFY H5 MODEL
# =====================================================================

print("\n" + "=" * 80)
print("VERIFYING SAVED MODEL")
print("=" * 80)


loaded_model = tf.keras.models.load_model(
    H5_PATH
)


test_sample = X_test_scaled[
    :5
]


verification_predictions = (

    loaded_model.predict(

        test_sample,

        verbose=0

    )

    .flatten()

)


print(
    "\nVerification predictions:"
)


for index, probability in enumerate(

    verification_predictions,

    start=1

):

    predicted_class = (

        "Anomaly"

        if probability >= 0.5

        else "Clean"

    )


    print(

        f"Sample {index}: "
        f"{probability:.4f} "
        f"-> {predicted_class}"

    )


# =====================================================================
# 40. DISPLAY GENERATED FILES
# =====================================================================

print("\n" + "=" * 80)
print("ALL FILES GENERATED SUCCESSFULLY")
print("=" * 80)


generated_files = [

    H5_PATH,

    PKL_PATH,

    YAML_PATH,

    JSON_PATH,

    ACCURACY_GRAPH_PATH,

    HEATMAP_PATH,

    COMPARISON_GRAPH_PATH,

    RESULT_CSV_PATH,

    RESULT_GRAPH_PATH,

    PREDICTION_CSV_PATH,

    PREDICTION_GRAPH_PATH,

    LABELED_CSV_PATH

]


for number, path in enumerate(

    generated_files,

    start=1

):

    if path.exists():

        status = "SUCCESS"

    else:

        status = "FAILED"


    print(

        f"{number:02d}. "
        f"{path.name:45s} "
        f"[{status}]"

    )


# =====================================================================
# 41. FINAL RESULT
# =====================================================================

print("\n" + "=" * 80)
print("FINAL SUMMARY")
print("=" * 80)


print(
    f"\nModel Accuracy : "
    f"{accuracy * 100:.2f}%"
)

print(
    f"Precision      : "
    f"{precision * 100:.2f}%"
)

print(
    f"Recall         : "
    f"{recall * 100:.2f}%"
)

print(
    f"F1 Score       : "
    f"{f1 * 100:.2f}%"
)

print(
    f"ROC-AUC        : "
    f"{roc_auc * 100:.2f}%"
)


print(
    "\nTotal Transactions:",
    total_records
)

print(
    "Clean Transactions:",
    clean_records
)

print(
    "Wrong Transactions:",
    anomaly_records
)


print(
    "\nAll graphs have been SHOWN and SAVED."
)


print(
    "\nOutput folder:"
)

print(
    PROJECT_DIR
)


print("\nGenerated visualization files:")

print(
    "1. accuracy_graph.png"
)

print(
    "2. heatmap.png"
)

print(
    "3. comparison_graph.png"
)

print(
    "4. result_graph.png"
)

print(
    "5. prediction_graph.png"
)


print("\nGenerated CSV files:")

print(
    "1. result.csv"
)

print(
    "2. prediction.csv"
)


print(
    "\nProject execution completed successfully."
)

print("=" * 80)
```
