```python
import os
import json
import random
import warnings
from pathlib import Path

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

import joblib
import yaml

from sklearn.model_selection import (
    train_test_split,
    StratifiedShuffleSplit
)

from sklearn.preprocessing import StandardScaler

from sklearn.linear_model import LogisticRegression

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    confusion_matrix,
    classification_report
)

import tensorflow as tf

from tensorflow.keras.models import Sequential

from tensorflow.keras.layers import (
    Dense,
    Dropout,
    BatchNormalization
)

from tensorflow.keras.callbacks import (
    EarlyStopping,
    ReduceLROnPlateau
)


# =====================================================================
# 2. SETTINGS
# =====================================================================

warnings.filterwarnings("ignore")

SEED = 42

random.seed(SEED)
np.random.seed(SEED)
tf.random.set_seed(SEED)


# =====================================================================
# 3. PROJECT PATH
# =====================================================================

PROJECT_DIR = Path(
    r"C:\Users\sagni\Downloads\Financial Transaction Data Quality"
)

CSV_PATH = (
    PROJECT_DIR
    / "dirty_financial_transactions.csv"
)

PROJECT_DIR.mkdir(
    parents=True,
    exist_ok=True
)


# =====================================================================
# 4. OUTPUT PATHS
# =====================================================================

H5_PATH = (
    PROJECT_DIR
    / "pso_financial_data_quality_model.h5"
)

PKL_PATH = (
    PROJECT_DIR
    / "pso_financial_data_quality_pipeline.pkl"
)

YAML_PATH = (
    PROJECT_DIR
    / "pso_financial_data_quality_config.yaml"
)

JSON_PATH = (
    PROJECT_DIR
    / "pso_financial_data_quality_results.json"
)


# Graphs

ACCURACY_GRAPH_PATH = (
    PROJECT_DIR
    / "pso_accuracy_graph.png"
)

HEATMAP_PATH = (
    PROJECT_DIR
    / "pso_heatmap.png"
)

COMPARISON_GRAPH_PATH = (
    PROJECT_DIR
    / "pso_comparison_graph.png"
)

RESULT_GRAPH_PATH = (
    PROJECT_DIR
    / "pso_result_graph.png"
)

PREDICTION_GRAPH_PATH = (
    PROJECT_DIR
    / "pso_prediction_graph.png"
)

FEATURE_SELECTION_GRAPH_PATH = (
    PROJECT_DIR
    / "pso_feature_selection_graph.png"
)

PSO_HISTORY_GRAPH_PATH = (
    PROJECT_DIR
    / "pso_optimization_history_graph.png"
)


# CSV files

RESULT_CSV_PATH = (
    PROJECT_DIR
    / "pso_result.csv"
)

PREDICTION_CSV_PATH = (
    PROJECT_DIR
    / "pso_prediction.csv"
)

FEATURE_SELECTION_CSV_PATH = (
    PROJECT_DIR
    / "pso_feature_selection.csv"
)

PSO_HISTORY_CSV_PATH = (
    PROJECT_DIR
    / "pso_optimization_history.csv"
)

LABELED_CSV_PATH = (
    PROJECT_DIR
    / "pso_labeled_dataset.csv"
)


# =====================================================================
# 5. LOAD DATASET
# =====================================================================

print("\n" + "=" * 80)
print("PSO-BASED FINANCIAL TRANSACTION DATA QUALITY PREDICTION")
print("=" * 80)


if not CSV_PATH.exists():

    raise FileNotFoundError(
        f"Dataset not found:\n{CSV_PATH}"
    )


print("\nLoading dataset...")


df = pd.read_csv(
    CSV_PATH,
    low_memory=False
)


original_df = df.copy()


print(
    "\nDataset loaded successfully."
)

print(
    "Dataset shape:",
    df.shape
)


# =====================================================================
# 6. VALIDATE COLUMNS
# =====================================================================

required_columns = [

    "Transaction_ID",
    "Transaction_Date",
    "Customer_ID",
    "Product_Name",
    "Quantity",
    "Price",
    "Payment_Method",
    "Transaction_Status"

]


missing_columns = [

    column

    for column in required_columns

    if column not in df.columns

]


if missing_columns:

    raise ValueError(
        f"Missing columns: {missing_columns}"
    )


# =====================================================================
# 7. HELPER FUNCTIONS
# =====================================================================


def safe_string(value):

    if pd.isna(value):

        return ""

    return str(value).strip()



def clean_price(value):

    if pd.isna(value):

        return np.nan

    value = str(value)

    value = value.replace(
        "$",
        ""
    )

    value = value.replace(
        ",",
        ""
    )

    value = value.strip()

    try:

        return float(value)

    except:

        return np.nan



def valid_transaction_id(value):

    value = safe_string(
        value
    )

    if value == "":

        return False

    if not value.startswith("T"):

        return False

    return value[1:].isdigit()



def valid_customer_id(value):

    value = safe_string(
        value
    )

    return value != ""


# =====================================================================
# 8. VALID CATEGORY VALUES
# =====================================================================

VALID_PRODUCTS = {

    "Laptop",
    "Tablet",
    "Smartphone",
    "Headphones",
    "Coffee Machine"

}


VALID_PAYMENT_METHODS = {

    "Credit Card",
    "PayPal",
    "Cash"

}


VALID_STATUSES = {

    "Completed",
    "Pending",
    "Failed"

}


# =====================================================================
# 9. BASIC DATA CLEANING
# =====================================================================

df["Price_Numeric"] = (

    df["Price"]
    .apply(
        clean_price
    )

)


df["Quantity_Numeric"] = pd.to_numeric(

    df["Quantity"],

    errors="coerce"

)


df["Parsed_Date"] = pd.to_datetime(

    df["Transaction_Date"],

    errors="coerce"

)


product_original = (

    df["Product_Name"]
    .fillna("")
    .astype(str)
    .str.strip()

)


payment_original = (

    df["Payment_Method"]
    .fillna("")
    .astype(str)
    .str.strip()

)


status_original = (

    df["Transaction_Status"]
    .fillna("")
    .astype(str)
    .str.strip()

)


# =====================================================================
# 10. CREATE ERROR RULES
# =====================================================================


# Transaction ID

df["Error_Missing_Transaction_ID"] = (

    df["Transaction_ID"]
    .isna()
    .astype(int)

)


df["Error_Invalid_Transaction_ID"] = (

    ~df["Transaction_ID"]
    .apply(
        valid_transaction_id
    )

).astype(int)


df["Error_Duplicate_Transaction_ID"] = (

    df["Transaction_ID"]
    .duplicated(
        keep=False
    )

    &

    df["Transaction_ID"]
    .notna()

).astype(int)


# Customer

df["Error_Missing_Customer_ID"] = (

    df["Customer_ID"]
    .isna()
    .astype(int)

)


df["Error_Invalid_Customer_ID"] = (

    ~df["Customer_ID"]
    .apply(
        valid_customer_id
    )

).astype(int)


# Date

df["Error_Missing_Date"] = (

    df["Transaction_Date"]
    .isna()
    .astype(int)

)


df["Error_Invalid_Date"] = (

    df["Transaction_Date"]
    .notna()

    &

    df["Parsed_Date"]
    .isna()

).astype(int)


# Product

df["Error_Missing_Product"] = (

    df["Product_Name"]
    .isna()
    .astype(int)

)


df["Error_Invalid_Product"] = (

    df["Product_Name"]
    .notna()

    &

    ~product_original.isin(
        VALID_PRODUCTS
    )

).astype(int)


# Quantity

df["Error_Missing_Quantity"] = (

    df["Quantity_Numeric"]
    .isna()
    .astype(int)

)


df["Error_Invalid_Quantity"] = (

    df["Quantity_Numeric"]
    .notna()

    &

    (
        df["Quantity_Numeric"]
        <= 0
    )

).astype(int)


# Price

df["Error_Missing_Price"] = (

    df["Price"]
    .isna()
    .astype(int)

)


df["Error_Invalid_Price_Format"] = (

    df["Price"]
    .notna()

    &

    df["Price_Numeric"]
    .isna()

).astype(int)


df["Error_Invalid_Price_Value"] = (

    df["Price_Numeric"]
    .notna()

    &

    (
        df["Price_Numeric"]
        <= 0
    )

).astype(int)


# Payment

df["Error_Missing_Payment_Method"] = (

    df["Payment_Method"]
    .isna()
    .astype(int)

)


df["Error_Nonstandard_Payment"] = (

    df["Payment_Method"]
    .notna()

    &

    ~payment_original.isin(
        VALID_PAYMENT_METHODS
    )

).astype(int)


# Status

df["Error_Missing_Status"] = (

    df["Transaction_Status"]
    .isna()
    .astype(int)

)


df["Error_Nonstandard_Status"] = (

    df["Transaction_Status"]
    .notna()

    &

    ~status_original.isin(
        VALID_STATUSES
    )

).astype(int)


# =====================================================================
# 11. ERROR DEFINITIONS
# =====================================================================

ERROR_COLUMNS = [

    "Error_Missing_Transaction_ID",
    "Error_Invalid_Transaction_ID",
    "Error_Duplicate_Transaction_ID",

    "Error_Missing_Customer_ID",
    "Error_Invalid_Customer_ID",

    "Error_Missing_Date",
    "Error_Invalid_Date",

    "Error_Missing_Product",
    "Error_Invalid_Product",

    "Error_Missing_Quantity",
    "Error_Invalid_Quantity",

    "Error_Missing_Price",
    "Error_Invalid_Price_Format",
    "Error_Invalid_Price_Value",

    "Error_Missing_Payment_Method",
    "Error_Nonstandard_Payment",

    "Error_Missing_Status",
    "Error_Nonstandard_Status"

]


ERROR_NAMES = {

    "Error_Missing_Transaction_ID":
        "Missing Transaction ID",

    "Error_Invalid_Transaction_ID":
        "Invalid Transaction ID",

    "Error_Duplicate_Transaction_ID":
        "Duplicate Transaction ID",

    "Error_Missing_Customer_ID":
        "Missing Customer ID",

    "Error_Invalid_Customer_ID":
        "Invalid Customer ID",

    "Error_Missing_Date":
        "Missing Date",

    "Error_Invalid_Date":
        "Invalid Date",

    "Error_Missing_Product":
        "Missing Product",

    "Error_Invalid_Product":
        "Invalid Product",

    "Error_Missing_Quantity":
        "Missing Quantity",

    "Error_Invalid_Quantity":
        "Invalid Quantity",

    "Error_Missing_Price":
        "Missing Price",

    "Error_Invalid_Price_Format":
        "Invalid Price Format",

    "Error_Invalid_Price_Value":
        "Invalid Price Value",

    "Error_Missing_Payment_Method":
        "Missing Payment Method",

    "Error_Nonstandard_Payment":
        "Nonstandard Payment Method",

    "Error_Missing_Status":
        "Missing Status",

    "Error_Nonstandard_Status":
        "Nonstandard Status"

}


# =====================================================================
# 12. CREATE TARGET
# =====================================================================

df["Error_Count"] = (

    df[
        ERROR_COLUMNS
    ]
    .sum(
        axis=1
    )

)


df["Is_Anomaly"] = (

    df["Error_Count"] > 0

).astype(int)


# =====================================================================
# 13. CREATE ERROR DESCRIPTION
# =====================================================================


def create_error_type(row):

    errors = []

    for column in ERROR_COLUMNS:

        if row[column] == 1:

            errors.append(
                ERROR_NAMES[
                    column
                ]
            )

    if len(errors) == 0:

        return "Clean"

    return " | ".join(
        errors
    )


df["Error_Type"] = df.apply(

    create_error_type,

    axis=1

)


# =====================================================================
# 14. SUMMARY
# =====================================================================

TOTAL_RECORDS = len(df)


CLEAN_RECORDS = int(

    (
        df["Is_Anomaly"] == 0
    ).sum()

)


ANOMALY_RECORDS = int(

    (
        df["Is_Anomaly"] == 1
    ).sum()

)


print("\n" + "=" * 80)
print("DATA QUALITY SUMMARY")
print("=" * 80)


print(
    "\nTotal records:",
    TOTAL_RECORDS
)

print(
    "Clean records:",
    CLEAN_RECORDS
)

print(
    "Anomaly records:",
    ANOMALY_RECORDS
)


# =====================================================================
# 15. FEATURE ENGINEERING
# =====================================================================

features = pd.DataFrame(
    index=df.index
)


# Transaction ID

features[
    "transaction_id_missing"
] = (

    df["Transaction_ID"]
    .isna()
    .astype(int)

)


features[
    "transaction_id_length"
] = (

    df["Transaction_ID"]
    .fillna("")
    .astype(str)
    .str.len()

)


features[
    "transaction_id_valid"
] = (

    df["Transaction_ID"]
    .apply(
        valid_transaction_id
    )
    .astype(int)

)


features[
    "transaction_id_duplicate"
] = (

    df["Transaction_ID"]
    .duplicated(
        keep=False
    )
    .astype(int)

)


# Customer

features[
    "customer_missing"
] = (

    df["Customer_ID"]
    .isna()
    .astype(int)

)


features[
    "customer_id_length"
] = (

    df["Customer_ID"]
    .fillna("")
    .astype(str)
    .str.len()

)


# Date

features[
    "date_missing"
] = (

    df["Transaction_Date"]
    .isna()
    .astype(int)

)


features[
    "date_valid"
] = (

    df["Parsed_Date"]
    .notna()
    .astype(int)

)


features[
    "year"
] = (

    df["Parsed_Date"]
    .dt.year
    .fillna(0)

)


features[
    "month"
] = (

    df["Parsed_Date"]
    .dt.month
    .fillna(0)

)


features[
    "day"
] = (

    df["Parsed_Date"]
    .dt.day
    .fillna(0)

)


# Product

features[
    "product_missing"
] = (

    df["Product_Name"]
    .isna()
    .astype(int)

)


features[
    "product_length"
] = (

    product_original
    .str.len()

)


features[
    "product_valid"
] = (

    product_original
    .isin(
        VALID_PRODUCTS
    )
    .astype(int)

)


# Quantity

features[
    "quantity_missing"
] = (

    df["Quantity_Numeric"]
    .isna()
    .astype(int)

)


features[
    "quantity_value"
] = (

    df["Quantity_Numeric"]
    .fillna(0)

)


features[
    "quantity_positive"
] = (

    df["Quantity_Numeric"]
    .gt(0)
    .fillna(False)
    .astype(int)

)


features[
    "quantity_negative"
] = (

    df["Quantity_Numeric"]
    .lt(0)
    .fillna(False)
    .astype(int)

)


# Price

features[
    "price_missing"
] = (

    df["Price"]
    .isna()
    .astype(int)

)


features[
    "price_value"
] = (

    df["Price_Numeric"]
    .fillna(0)

)


features[
    "price_positive"
] = (

    df["Price_Numeric"]
    .gt(0)
    .fillna(False)
    .astype(int)

)


features[
    "price_negative"
] = (

    df["Price_Numeric"]
    .lt(0)
    .fillna(False)
    .astype(int)

)


features[
    "price_has_dollar"
] = (

    df["Price"]
    .fillna("")
    .astype(str)
    .str.contains(
        r"\$",
        regex=True
    )
    .astype(int)

)


features[
    "price_length"
] = (

    df["Price"]
    .fillna("")
    .astype(str)
    .str.len()

)


# Payment

features[
    "payment_missing"
] = (

    df["Payment_Method"]
    .isna()
    .astype(int)

)


features[
    "payment_valid"
] = (

    payment_original
    .isin(
        VALID_PAYMENT_METHODS
    )
    .astype(int)

)


features[
    "payment_length"
] = (

    payment_original
    .str.len()

)


# Status

features[
    "status_missing"
] = (

    df["Transaction_Status"]
    .isna()
    .astype(int)

)


features[
    "status_valid"
] = (

    status_original
    .isin(
        VALID_STATUSES
    )
    .astype(int)

)


features[
    "status_length"
] = (

    status_original
    .str.len()

)


# =====================================================================
# 16. CLEAN FEATURES
# =====================================================================

features = features.replace(

    [
        np.inf,
        -np.inf
    ],

    np.nan

)


features = features.fillna(
    0
)


features = features.astype(
    np.float32
)


X = features.copy()


y = (

    df["Is_Anomaly"]
    .astype(
        np.int32
    )

)


NUMBER_OF_FEATURES = (
    X.shape[1]
)


print(
    "\nGenerated ML features:",
    NUMBER_OF_FEATURES
)


# =====================================================================
# 17. MAIN TRAIN / TEST SPLIT
# =====================================================================

X_train_full, X_test, y_train_full, y_test = train_test_split(

    X,
    y,

    test_size=0.20,

    random_state=SEED,

    stratify=y

)


print(
    "\nTraining records:",
    len(
        X_train_full
    )
)

print(
    "Testing records:",
    len(
        X_test
    )
)


# =====================================================================
# 18. CREATE PSO OPTIMIZATION DATASET
# =====================================================================

PSO_SAMPLE_SIZE = min(

    30000,

    len(
        X_train_full
    )

)


splitter = StratifiedShuffleSplit(

    n_splits=1,

    train_size=PSO_SAMPLE_SIZE,

    random_state=SEED

)


pso_indices, _ = next(

    splitter.split(

        X_train_full,

        y_train_full

    )

)


X_pso = (

    X_train_full
    .iloc[
        pso_indices
    ]
    .reset_index(
        drop=True
    )

)


y_pso = (

    y_train_full
    .iloc[
        pso_indices
    ]
    .reset_index(
        drop=True
    )

)


X_pso_train, X_pso_val, y_pso_train, y_pso_val = train_test_split(

    X_pso,
    y_pso,

    test_size=0.25,

    random_state=SEED,

    stratify=y_pso

)


# =====================================================================
# 19. PSO CONFIGURATION
# =====================================================================
#
# Binary Particle Swarm Optimization
#
# Each particle position determines which features are selected.
#
# Sigmoid(velocity) is used to convert velocity into probability.
#
# =====================================================================

PSO_PARTICLES = 20

PSO_ITERATIONS = 15


# Inertia

PSO_W = 0.729


# Cognitive coefficient

PSO_C1 = 1.49445


# Social coefficient

PSO_C2 = 1.49445


print("\n" + "=" * 80)
print("PARTICLE SWARM OPTIMIZATION")
print("=" * 80)


print(
    "\nParticles:",
    PSO_PARTICLES
)

print(
    "Iterations:",
    PSO_ITERATIONS
)

print(
    "Features:",
    NUMBER_OF_FEATURES
)


# =====================================================================
# 20. PSO FITNESS FUNCTION
# =====================================================================


def evaluate_particle(
    particle_position
):

    selected_indices = np.where(

        particle_position == 1

    )[0]


    # No features selected

    if len(selected_indices) == 0:

        return 0.0


    X_train_selected = (

        X_pso_train
        .iloc[
            :,
            selected_indices
        ]

    )


    X_val_selected = (

        X_pso_val
        .iloc[
            :,
            selected_indices
        ]

    )


    temp_scaler = StandardScaler()


    X_train_scaled = temp_scaler.fit_transform(

        X_train_selected

    )


    X_val_scaled = temp_scaler.transform(

        X_val_selected

    )


    temp_model = LogisticRegression(

        max_iter=500,

        random_state=SEED,

        class_weight="balanced",

        solver="liblinear"

    )


    try:

        temp_model.fit(

            X_train_scaled,

            y_pso_train

        )


        predictions = temp_model.predict(

            X_val_scaled

        )


        model_f1 = f1_score(

            y_pso_val,

            predictions,

            zero_division=0

        )


    except:

        return 0.0


    # ---------------------------------------------------------
    # Feature reduction component
    # ---------------------------------------------------------

    selected_ratio = (

        len(
            selected_indices
        )

        /

        NUMBER_OF_FEATURES

    )


    feature_reduction = (

        1.0
        -
        selected_ratio

    )


    # ---------------------------------------------------------
    # Fitness
    #
    # 95% classification performance
    # 5% feature reduction
    # ---------------------------------------------------------

    fitness = (

        0.95
        *
        model_f1

        +

        0.05
        *
        feature_reduction

    )


    return float(
        fitness
    )


# =====================================================================
# 21. INITIALIZE PARTICLES
# =====================================================================

positions = np.random.randint(

    0,
    2,

    size=(

        PSO_PARTICLES,

        NUMBER_OF_FEATURES

    )

)


# Ensure every particle has at least one feature

for particle_index in range(

    PSO_PARTICLES

):

    if np.sum(
        positions[
            particle_index
        ]
    ) == 0:

        positions[
            particle_index,
            np.random.randint(
                0,
                NUMBER_OF_FEATURES
            )
        ] = 1


# First particle contains every feature

positions[
    0
] = np.ones(

    NUMBER_OF_FEATURES,

    dtype=int

)


# Initialize velocities

velocities = np.random.uniform(

    low=-1,

    high=1,

    size=(

        PSO_PARTICLES,

        NUMBER_OF_FEATURES

    )

)


# =====================================================================
# 22. INITIALIZE PERSONAL BEST
# =====================================================================

personal_best_positions = positions.copy()


personal_best_fitness = np.zeros(

    PSO_PARTICLES,

    dtype=float

)


for particle_index in range(

    PSO_PARTICLES

):

    personal_best_fitness[
        particle_index
    ] = evaluate_particle(

        positions[
            particle_index
        ]

    )


# =====================================================================
# 23. INITIALIZE GLOBAL BEST
# =====================================================================

best_particle_index = np.argmax(

    personal_best_fitness

)


global_best_position = (

    personal_best_positions[
        best_particle_index
    ].copy()

)


global_best_fitness = float(

    personal_best_fitness[
        best_particle_index
    ]

)


print(
    "\nInitial best fitness:",
    global_best_fitness
)


# =====================================================================
# 24. PSO OPTIMIZATION
# =====================================================================

pso_history = []


for iteration in range(

    PSO_ITERATIONS

):


    print(
        f"\nPSO Iteration "
        f"{iteration + 1}/"
        f"{PSO_ITERATIONS}"
    )


    # ---------------------------------------------------------
    # Process each particle
    # ---------------------------------------------------------

    for particle_index in range(

        PSO_PARTICLES

    ):


        # Random coefficients

        r1 = np.random.rand(
            NUMBER_OF_FEATURES
        )

        r2 = np.random.rand(
            NUMBER_OF_FEATURES
        )


        # -----------------------------------------------------
        # Update velocity
        # -----------------------------------------------------

        cognitive_component = (

            PSO_C1
            *
            r1
            *
            (
                personal_best_positions[
                    particle_index
                ]

                -

                positions[
                    particle_index
                ]
            )

        )


        social_component = (

            PSO_C2
            *
            r2
            *
            (
                global_best_position

                -

                positions[
                    particle_index
                ]
            )

        )


        velocities[
            particle_index
        ] = (

            PSO_W
            *
            velocities[
                particle_index
            ]

            +

            cognitive_component

            +

            social_component

        )


        # Clip extreme velocities

        velocities[
            particle_index
        ] = np.clip(

            velocities[
                particle_index
            ],

            -6,
            6

        )


        # -----------------------------------------------------
        # Convert velocity to probability
        # -----------------------------------------------------

        probabilities = (

            1

            /

            (
                1

                +

                np.exp(
                    -velocities[
                        particle_index
                    ]
                )
            )

        )


        random_values = np.random.rand(

            NUMBER_OF_FEATURES

        )


        # Binary PSO update

        positions[
            particle_index
        ] = (

            random_values
            <
            probabilities

        ).astype(int)


        # Ensure at least one feature

        if np.sum(

            positions[
                particle_index
            ]

        ) == 0:


            positions[
                particle_index,
                np.random.randint(
                    0,
                    NUMBER_OF_FEATURES
                )
            ] = 1


        # -----------------------------------------------------
        # Evaluate new position
        # -----------------------------------------------------

        current_fitness = evaluate_particle(

            positions[
                particle_index
            ]

        )


        # -----------------------------------------------------
        # Update personal best
        # -----------------------------------------------------

        if (

            current_fitness

            >

            personal_best_fitness[
                particle_index
            ]

        ):


            personal_best_fitness[
                particle_index
            ] = current_fitness


            personal_best_positions[
                particle_index
            ] = positions[
                particle_index
            ].copy()


        # -----------------------------------------------------
        # Update global best
        # -----------------------------------------------------

        if (

            current_fitness

            >

            global_best_fitness

        ):


            global_best_fitness = float(

                current_fitness

            )


            global_best_position = (

                positions[
                    particle_index
                ]
                .copy()

            )


    # ---------------------------------------------------------
    # End of iteration
    # ---------------------------------------------------------

    selected_features_count = int(

        np.sum(
            global_best_position
        )

    )


    mean_fitness = float(

        np.mean(
            personal_best_fitness
        )

    )


    print(
        "Global best fitness:",
        round(
            global_best_fitness,
            6
        )
    )


    print(
        "Selected features:",
        selected_features_count,
        "/",
        NUMBER_OF_FEATURES
    )


    pso_history.append(

        {

            "Iteration":
                iteration + 1,

            "Global_Best_Fitness":
                global_best_fitness,

            "Mean_Personal_Best_Fitness":
                mean_fitness,

            "Selected_Features":
                selected_features_count

        }

    )


# =====================================================================
# 25. FINAL PSO SELECTED FEATURES
# =====================================================================

selected_indices = np.where(

    global_best_position == 1

)[0]


selected_features = [

    X.columns[
        index
    ]

    for index in selected_indices

]


print("\n" + "=" * 80)
print("PSO FEATURE SELECTION COMPLETED")
print("=" * 80)


print(
    "\nBest PSO fitness:",
    global_best_fitness
)


print(
    "Original feature count:",
    NUMBER_OF_FEATURES
)


print(
    "Selected feature count:",
    len(
        selected_features
    )
)


print(
    "\nSelected features:"
)


for feature in selected_features:

    print(
        " -",
        feature
    )


# =====================================================================
# 26. SAVE PSO OPTIMIZATION HISTORY
# =====================================================================

pso_history_df = pd.DataFrame(

    pso_history

)


pso_history_df.to_csv(

    PSO_HISTORY_CSV_PATH,

    index=False

)


print(
    "\nSaved:",
    PSO_HISTORY_CSV_PATH
)


# =====================================================================
# 27. PSO OPTIMIZATION GRAPH
# =====================================================================

plt.figure(
    figsize=(10, 6)
)


plt.plot(

    pso_history_df[
        "Iteration"
    ],

    pso_history_df[
        "Global_Best_Fitness"
    ],

    marker="o",

    label="Global Best Fitness"

)


plt.plot(

    pso_history_df[
        "Iteration"
    ],

    pso_history_df[
        "Mean_Personal_Best_Fitness"
    ],

    marker="s",

    label="Mean Personal Best Fitness"

)


plt.xlabel(
    "PSO Iteration"
)

plt.ylabel(
    "Fitness"
)

plt.title(
    "PSO Optimization Convergence"
)

plt.legend()

plt.grid(
    True,
    alpha=0.3
)

plt.tight_layout()


plt.savefig(

    PSO_HISTORY_GRAPH_PATH,

    dpi=300,

    bbox_inches="tight"

)


print(
    "Saved:",
    PSO_HISTORY_GRAPH_PATH
)


plt.show()

plt.close()


# =====================================================================
# 28. SAVE FEATURE SELECTION CSV
# =====================================================================

feature_selection_df = pd.DataFrame(

    {

        "Feature":

            list(
                X.columns
            ),

        "PSO_Selected":

            [

                1
                if feature in selected_features
                else 0

                for feature
                in X.columns

            ]

    }

)


feature_selection_df.to_csv(

    FEATURE_SELECTION_CSV_PATH,

    index=False

)


print(
    "Saved:",
    FEATURE_SELECTION_CSV_PATH
)


# =====================================================================
# 29. PSO FEATURE SELECTION GRAPH
# =====================================================================

plt.figure(
    figsize=(12, 8)
)


plt.barh(

    feature_selection_df[
        "Feature"
    ],

    feature_selection_df[
        "PSO_Selected"
    ]

)


plt.xlabel(
    "PSO Selected (1 = Yes, 0 = No)"
)

plt.ylabel(
    "Feature"
)

plt.title(
    "PSO Feature Selection Result"
)

plt.tight_layout()


plt.savefig(

    FEATURE_SELECTION_GRAPH_PATH,

    dpi=300,

    bbox_inches="tight"

)


print(
    "Saved:",
    FEATURE_SELECTION_GRAPH_PATH
)


plt.show()

plt.close()


# =====================================================================
# 30. SELECT PSO FEATURES FOR FINAL MODEL
# =====================================================================

X_train = (

    X_train_full[
        selected_features
    ]
    .copy()

)


X_test_selected = (

    X_test[
        selected_features
    ]
    .copy()

)


# =====================================================================
# 31. SCALE FINAL DATA
# =====================================================================

scaler = StandardScaler()


X_train_scaled = scaler.fit_transform(

    X_train

)


X_test_scaled = scaler.transform(

    X_test_selected

)


# =====================================================================
# 32. CALCULATE CLASS WEIGHTS
# =====================================================================

class_counts = np.bincount(

    y_train_full

)


class_weight = None


if len(
    class_counts
) == 2:


    total_training_records = len(

        y_train_full

    )


    class_weight = {

        0:

            total_training_records

            /

            (
                2
                *
                class_counts[0]
            ),


        1:

            total_training_records

            /

            (
                2
                *
                class_counts[1]
            )

    }


print(
    "\nClass weights:",
    class_weight
)


# =====================================================================
# 33. BUILD FINAL PSO-ANN MODEL
# =====================================================================

model = Sequential(

    [

        tf.keras.Input(

            shape=(

                X_train_scaled.shape[1],

            )

        ),


        Dense(

            128,

            activation="relu"

        ),


        BatchNormalization(),


        Dropout(
            0.30
        ),


        Dense(

            64,

            activation="relu"

        ),


        BatchNormalization(),


        Dropout(
            0.25
        ),


        Dense(

            32,

            activation="relu"

        ),


        Dropout(
            0.20
        ),


        Dense(

            16,

            activation="relu"

        ),


        Dense(

            1,

            activation="sigmoid"

        )

    ]

)


# =====================================================================
# 34. COMPILE MODEL
# =====================================================================

model.compile(

    optimizer=tf.keras.optimizers.Adam(

        learning_rate=0.001

    ),

    loss="binary_crossentropy",

    metrics=[

        "accuracy",

        tf.keras.metrics.Precision(
            name="precision"
        ),

        tf.keras.metrics.Recall(
            name="recall"
        ),

        tf.keras.metrics.AUC(
            name="auc"
        )

    ]

)


model.summary()


# =====================================================================
# 35. CALLBACKS
# =====================================================================

early_stopping = EarlyStopping(

    monitor="val_loss",

    patience=7,

    restore_best_weights=True,

    verbose=1

)


reduce_lr = ReduceLROnPlateau(

    monitor="val_loss",

    factor=0.5,

    patience=3,

    min_lr=1e-6,

    verbose=1

)


# =====================================================================
# 36. TRAIN FINAL PSO-ANN MODEL
# =====================================================================

print("\n" + "=" * 80)
print("TRAINING PSO-OPTIMIZED ANN")
print("=" * 80)


history = model.fit(

    X_train_scaled,
    y_train_full,

    validation_split=0.20,

    epochs=50,

    batch_size=256,

    callbacks=[

        early_stopping,
        reduce_lr

    ],

    class_weight=class_weight,

    verbose=1

)


# =====================================================================
# 37. PREDICTIONS
# =====================================================================

prediction_probability = (

    model.predict(

        X_test_scaled,

        verbose=0

    )

    .flatten()

)


prediction_label = (

    prediction_probability

    >= 0.50

).astype(int)


# =====================================================================
# 38. METRICS
# =====================================================================

accuracy = accuracy_score(

    y_test,

    prediction_label

)


precision = precision_score(

    y_test,

    prediction_label,

    zero_division=0

)


recall = recall_score(

    y_test,

    prediction_label,

    zero_division=0

)


f1 = f1_score(

    y_test,

    prediction_label,

    zero_division=0

)


try:

    roc_auc = roc_auc_score(

        y_test,

        prediction_probability

    )


except:

    roc_auc = 0.0


cm = confusion_matrix(

    y_test,

    prediction_label

)


classification_results = classification_report(

    y_test,

    prediction_label,

    target_names=[

        "Clean",
        "Anomaly"

    ],

    output_dict=True,

    zero_division=0

)


print("\n" + "=" * 80)
print("PSO MODEL RESULTS")
print("=" * 80)


print(
    f"\nAccuracy : {accuracy * 100:.2f}%"
)

print(
    f"Precision: {precision * 100:.2f}%"
)

print(
    f"Recall   : {recall * 100:.2f}%"
)

print(
    f"F1 Score : {f1 * 100:.2f}%"
)

print(
    f"ROC-AUC  : {roc_auc * 100:.2f}%"
)


print("\nConfusion Matrix:")

print(
    cm
)


print("\nClassification Report:")


print(

    classification_report(

        y_test,

        prediction_label,

        target_names=[

            "Clean",
            "Anomaly"

        ],

        zero_division=0

    )

)


# =====================================================================
# 39. PSO ACCURACY GRAPH
# =====================================================================

epochs_range = range(

    1,

    len(
        history.history[
            "accuracy"
        ]
    ) + 1

)


plt.figure(
    figsize=(10, 6)
)


plt.plot(

    epochs_range,

    history.history[
        "accuracy"
    ],

    marker="o",

    markersize=3,

    label="Training Accuracy"

)


plt.plot(

    epochs_range,

    history.history[
        "val_accuracy"
    ],

    marker="o",

    markersize=3,

    label="Validation Accuracy"

)


plt.xlabel(
    "Epoch"
)

plt.ylabel(
    "Accuracy"
)

plt.title(
    "PSO-Optimized ANN Accuracy"
)

plt.legend()

plt.grid(
    True,
    alpha=0.3
)

plt.tight_layout()


plt.savefig(

    ACCURACY_GRAPH_PATH,

    dpi=300,

    bbox_inches="tight"

)


print(
    "\nSaved:",
    ACCURACY_GRAPH_PATH
)


plt.show()

plt.close()


# =====================================================================
# 40. PSO CONFUSION MATRIX HEATMAP
# =====================================================================

plt.figure(
    figsize=(7, 6)
)


plt.imshow(
    cm
)


plt.title(
    "PSO Confusion Matrix Heatmap"
)

plt.xlabel(
    "Predicted Class"
)

plt.ylabel(
    "Actual Class"
)


plt.xticks(

    [0, 1],

    [
        "Clean",
        "Anomaly"
    ]

)


plt.yticks(

    [0, 1],

    [
        "Clean",
        "Anomaly"
    ]

)


for i in range(

    cm.shape[0]

):


    for j in range(

        cm.shape[1]

    ):


        plt.text(

            j,
            i,

            str(
                cm[i, j]
            ),

            ha="center",

            va="center",

            fontsize=14

        )


plt.colorbar(
    label="Number of Transactions"
)


plt.tight_layout()


plt.savefig(

    HEATMAP_PATH,

    dpi=300,

    bbox_inches="tight"

)


print(
    "Saved:",
    HEATMAP_PATH
)


plt.show()

plt.close()


# =====================================================================
# 41. PSO COMPARISON GRAPH
# =====================================================================

metric_names = [

    "Accuracy",
    "Precision",
    "Recall",
    "F1 Score",
    "ROC-AUC"

]


metric_values = [

    accuracy * 100,
    precision * 100,
    recall * 100,
    f1 * 100,
    roc_auc * 100

]


plt.figure(
    figsize=(10, 6)
)


bars = plt.bar(

    metric_names,

    metric_values

)


plt.xlabel(
    "Evaluation Metric"
)

plt.ylabel(
    "Score (%)"
)

plt.title(
    "PSO Model Performance Comparison"
)


plt.ylim(
    0,
    105
)


for bar, value in zip(

    bars,
    metric_values

):


    plt.text(

        bar.get_x()

        +

        bar.get_width() / 2,

        value + 1,

        f"{value:.2f}%",

        ha="center"

    )


plt.grid(

    axis="y",

    alpha=0.3

)


plt.tight_layout()


plt.savefig(

    COMPARISON_GRAPH_PATH,

    dpi=300,

    bbox_inches="tight"

)


print(
    "Saved:",
    COMPARISON_GRAPH_PATH
)


plt.show()

plt.close()


# =====================================================================
# 42. CREATE PSO RESULT CSV
# =====================================================================

result_records = []


for column in ERROR_COLUMNS:


    error_count = int(

        df[column]
        .sum()

    )


    percentage = (

        error_count

        /

        TOTAL_RECORDS

        *

        100

    )


    result_records.append(

        {

            "Error_Type":

                ERROR_NAMES[
                    column
                ],

            "Error_Count":

                error_count,

            "Percentage":

                round(
                    percentage,
                    4
                )

        }

    )


result_df = pd.DataFrame(

    result_records

)


result_df = result_df.sort_values(

    by="Error_Count",

    ascending=False

).reset_index(

    drop=True

)


result_df.to_csv(

    RESULT_CSV_PATH,

    index=False

)


print(
    "\nSaved:",
    RESULT_CSV_PATH
)


# =====================================================================
# 43. PSO RESULT GRAPH
# =====================================================================

plt.figure(
    figsize=(13, 8)
)


bars = plt.barh(

    result_df[
        "Error_Type"
    ],

    result_df[
        "Error_Count"
    ]

)


plt.xlabel(
    "Number of Records"
)

plt.ylabel(
    "Data Quality Error"
)

plt.title(
    "PSO Financial Data Quality Error Distribution"
)


plt.gca().invert_yaxis()


for bar in bars:


    value = bar.get_width()


    plt.text(

        value,

        bar.get_y()

        +

        bar.get_height() / 2,

        f" {int(value)}",

        va="center"

    )


plt.grid(

    axis="x",

    alpha=0.3

)


plt.tight_layout()


plt.savefig(

    RESULT_GRAPH_PATH,

    dpi=300,

    bbox_inches="tight"

)


print(
    "Saved:",
    RESULT_GRAPH_PATH
)


plt.show()

plt.close()


# =====================================================================
# 44. CREATE PSO PREDICTION CSV
# =====================================================================

prediction_df = (

    original_df

    .loc[
        X_test.index
    ]

    .copy()

)


prediction_df[
    "Actual_Label"
] = y_test.values


prediction_df[
    "Actual_Result"
] = np.where(

    y_test.values == 1,

    "Anomaly",

    "Clean"

)


prediction_df[
    "PSO_Prediction_Probability"
] = prediction_probability


prediction_df[
    "PSO_Predicted_Label"
] = prediction_label


prediction_df[
    "PSO_Predicted_Result"
] = np.where(

    prediction_label == 1,

    "Anomaly",

    "Clean"

)


prediction_df[
    "Correct_Prediction"
] = (

    y_test.values

    ==

    prediction_label

)


prediction_df[
    "Error_Count"
] = (

    df.loc[

        X_test.index,

        "Error_Count"

    ].values

)


prediction_df[
    "Error_Type"
] = (

    df.loc[

        X_test.index,

        "Error_Type"

    ].values

)


prediction_df.to_csv(

    PREDICTION_CSV_PATH,

    index=False

)


print(
    "Saved:",
    PREDICTION_CSV_PATH
)


# =====================================================================
# 45. PSO PREDICTION GRAPH
# =====================================================================

NUMBER_TO_PLOT = min(

    100,

    len(
        prediction_df
    )

)


plot_predictions = prediction_df.iloc[

    :NUMBER_TO_PLOT

]


x_positions = np.arange(

    NUMBER_TO_PLOT

)


plt.figure(
    figsize=(15, 7)
)


plt.plot(

    x_positions,

    plot_predictions[
        "Actual_Label"
    ],

    marker="o",

    markersize=4,

    label="Actual Label"

)


plt.plot(

    x_positions,

    plot_predictions[
        "PSO_Prediction_Probability"
    ],

    marker=".",

    markersize=4,

    label="PSO Prediction Probability"

)


plt.axhline(

    y=0.5,

    linestyle="--",

    label="Prediction Threshold"

)


plt.xlabel(
    "Test Transaction Sample"
)

plt.ylabel(
    "Prediction Value"
)

plt.title(
    "PSO Actual vs Predicted Wrongness"
)


plt.yticks(

    [
        0,
        0.5,
        1
    ],

    [
        "Clean (0)",
        "Threshold",
        "Anomaly (1)"
    ]

)


plt.legend()

plt.grid(
    True,
    alpha=0.3
)

plt.tight_layout()


plt.savefig(

    PREDICTION_GRAPH_PATH,

    dpi=300,

    bbox_inches="tight"

)


print(
    "Saved:",
    PREDICTION_GRAPH_PATH
)


plt.show()

plt.close()


# =====================================================================
# 46. SAVE PSO LABELED DATASET
# =====================================================================

labeled_df = original_df.copy()


for column in ERROR_COLUMNS:

    labeled_df[
        column
    ] = df[
        column
    ]


labeled_df[
    "Error_Count"
] = df[
    "Error_Count"
]


labeled_df[
    "Error_Type"
] = df[
    "Error_Type"
]


labeled_df[
    "Is_Anomaly"
] = df[
    "Is_Anomaly"
]


labeled_df.to_csv(

    LABELED_CSV_PATH,

    index=False

)


print(
    "\nSaved:",
    LABELED_CSV_PATH
)


# =====================================================================
# 47. SAVE H5 MODEL
# =====================================================================

model.save(

    H5_PATH

)


print(
    "Saved:",
    H5_PATH
)


# =====================================================================
# 48. SAVE PKL PIPELINE
# =====================================================================

pipeline_data = {

    "algorithm":

        "Particle Swarm Optimization + ANN",

    "pso_type":

        "Binary Particle Swarm Optimization",

    "scaler":

        scaler,

    "all_features":

        list(
            X.columns
        ),

    "selected_features":

        selected_features,

    "pso_best_position":

        global_best_position.tolist(),

    "pso_best_fitness":

        float(
            global_best_fitness
        ),

    "prediction_threshold":

        0.50,

    "valid_products":

        list(
            VALID_PRODUCTS
        ),

    "valid_payment_methods":

        list(
            VALID_PAYMENT_METHODS
        ),

    "valid_statuses":

        list(
            VALID_STATUSES
        ),

    "error_columns":

        ERROR_COLUMNS,

    "error_names":

        ERROR_NAMES

}


joblib.dump(

    pipeline_data,

    PKL_PATH

)


print(
    "Saved:",
    PKL_PATH
)


# =====================================================================
# 49. SAVE PSO YAML CONFIG
# =====================================================================

yaml_config = {

    "project": {

        "name":

            "PSO Financial Transaction Data Quality Prediction",

        "dataset":

            str(
                CSV_PATH
            ),

        "output_directory":

            str(
                PROJECT_DIR
            )

    },


    "dataset": {

        "total_records":

            int(
                TOTAL_RECORDS
            ),

        "clean_records":

            int(
                CLEAN_RECORDS
            ),

        "anomaly_records":

            int(
                ANOMALY_RECORDS
            )

    },


    "pso": {

        "algorithm":

            "Binary Particle Swarm Optimization",

        "particles":

            PSO_PARTICLES,

        "iterations":

            PSO_ITERATIONS,

        "inertia_weight":

            PSO_W,

        "cognitive_coefficient":

            PSO_C1,

        "social_coefficient":

            PSO_C2,

        "best_fitness":

            float(
                global_best_fitness
            ),

        "original_feature_count":

            int(
                NUMBER_OF_FEATURES
            ),

        "selected_feature_count":

            int(
                len(
                    selected_features
                )
            ),

        "selected_features":

            selected_features

    },


    "ann": {

        "hidden_layers": [

            128,
            64,
            32,
            16

        ],

        "optimizer":

            "Adam",

        "learning_rate":

            0.001,

        "loss":

            "binary_crossentropy",

        "batch_size":

            256,

        "maximum_epochs":

            50,

        "prediction_threshold":

            0.5

    },


    "metrics": {

        "accuracy":

            float(
                accuracy
            ),

        "precision":

            float(
                precision
            ),

        "recall":

            float(
                recall
            ),

        "f1_score":

            float(
                f1
            ),

        "roc_auc":

            float(
                roc_auc
            )

    },


    "output_files": {

        "h5_model":

            H5_PATH.name,

        "pkl_pipeline":

            PKL_PATH.name,

        "yaml_config":

            YAML_PATH.name,

        "json_results":

            JSON_PATH.name,

        "accuracy_graph":

            ACCURACY_GRAPH_PATH.name,

        "heatmap":

            HEATMAP_PATH.name,

        "comparison_graph":

            COMPARISON_GRAPH_PATH.name,

        "result_csv":

            RESULT_CSV_PATH.name,

        "result_graph":

            RESULT_GRAPH_PATH.name,

        "prediction_csv":

            PREDICTION_CSV_PATH.name,

        "prediction_graph":

            PREDICTION_GRAPH_PATH.name,

        "feature_selection_csv":

            FEATURE_SELECTION_CSV_PATH.name,

        "feature_selection_graph":

            FEATURE_SELECTION_GRAPH_PATH.name,

        "optimization_history_csv":

            PSO_HISTORY_CSV_PATH.name,

        "optimization_history_graph":

            PSO_HISTORY_GRAPH_PATH.name

    }

}


with open(

    YAML_PATH,

    "w",

    encoding="utf-8"

) as yaml_file:


    yaml.safe_dump(

        yaml_config,

        yaml_file,

        sort_keys=False,

        allow_unicode=True

    )


print(
    "Saved:",
    YAML_PATH
)


# =====================================================================
# 50. SAVE PSO JSON RESULTS
# =====================================================================

training_history = {}


for key, values in history.history.items():


    training_history[
        key
    ] = [

        float(
            value
        )

        for value in values

    ]


error_distribution = {

    ERROR_NAMES[
        column
    ]:

        int(
            df[column]
            .sum()
        )

    for column in ERROR_COLUMNS

}


json_results = {

    "project_name":

        "PSO Financial Transaction Data Quality Prediction",

    "optimization_algorithm":

        "Binary Particle Swarm Optimization",

    "classifier":

        "Artificial Neural Network",

    "total_records":

        int(
            TOTAL_RECORDS
        ),

    "clean_records":

        int(
            CLEAN_RECORDS
        ),

    "anomaly_records":

        int(
            ANOMALY_RECORDS
        ),

    "training_records":

        int(
            len(
                X_train_full
            )
        ),

    "testing_records":

        int(
            len(
                X_test
            )
        ),

    "pso": {

        "particles":

            PSO_PARTICLES,

        "iterations":

            PSO_ITERATIONS,

        "inertia":

            PSO_W,

        "c1":

            PSO_C1,

        "c2":

            PSO_C2,

        "best_fitness":

            float(
                global_best_fitness
            ),

        "best_position":

            global_best_position.tolist(),

        "original_feature_count":

            int(
                NUMBER_OF_FEATURES
            ),

        "selected_feature_count":

            int(
                len(
                    selected_features
                )
            ),

        "selected_features":

            selected_features

    },

    "metrics": {

        "accuracy":

            float(
                accuracy
            ),

        "precision":

            float(
                precision
            ),

        "recall":

            float(
                recall
            ),

        "f1_score":

            float(
                f1
            ),

        "roc_auc":

            float(
                roc_auc
            )

    },

    "confusion_matrix":

        cm.tolist(),

    "classification_report":

        classification_results,

    "error_distribution":

        error_distribution,

    "pso_optimization_history":

        pso_history,

    "training_history":

        training_history

}


with open(

    JSON_PATH,

    "w",

    encoding="utf-8"

) as json_file:


    json.dump(

        json_results,

        json_file,

        indent=4,

        ensure_ascii=False

    )


print(
    "Saved:",
    JSON_PATH
)


# =====================================================================
# 51. VERIFY SAVED MODEL
# =====================================================================

print("\n" + "=" * 80)
print("VERIFYING PSO MODEL")
print("=" * 80)


loaded_model = tf.keras.models.load_model(

    H5_PATH

)


verification_probabilities = (

    loaded_model.predict(

        X_test_scaled[
            :5
        ],

        verbose=0

    )

    .flatten()

)


for index, probability in enumerate(

    verification_probabilities,

    start=1

):


    predicted_result = (

        "ANOMALY"

        if probability >= 0.50

        else "CLEAN"

    )


    print(

        f"Sample {index}: "
        f"{probability:.4f} "
        f"-> {predicted_result}"

    )


# =====================================================================
# 52. VERIFY PKL
# =====================================================================

loaded_pipeline = joblib.load(

    PKL_PATH

)


print(
    "\nPKL loaded successfully."
)


print(
    "Stored selected features:",
    len(
        loaded_pipeline[
            "selected_features"
        ]
    )
)


# =====================================================================
# 53. CHECK GENERATED FILES
# =====================================================================

generated_files = [

    H5_PATH,

    PKL_PATH,

    YAML_PATH,

    JSON_PATH,

    ACCURACY_GRAPH_PATH,

    HEATMAP_PATH,

    COMPARISON_GRAPH_PATH,

    RESULT_CSV_PATH,

    RESULT_GRAPH_PATH,

    PREDICTION_CSV_PATH,

    PREDICTION_GRAPH_PATH,

    FEATURE_SELECTION_CSV_PATH,

    FEATURE_SELECTION_GRAPH_PATH,

    PSO_HISTORY_CSV_PATH,

    PSO_HISTORY_GRAPH_PATH,

    LABELED_CSV_PATH

]


print("\n" + "=" * 80)
print("GENERATED PSO FILES")
print("=" * 80)


for number, file_path in enumerate(

    generated_files,

    start=1

):


    status = (

        "SUCCESS"

        if file_path.exists()

        else "FAILED"

    )


    print(

        f"{number:02d}. "
        f"{file_path.name:48s} "
        f"[{status}]"

    )


# =====================================================================
# 54. FINAL SUMMARY
# =====================================================================

print("\n" + "=" * 80)
print("FINAL PSO RESULTS")
print("=" * 80)


print(
    "\nOptimization Algorithm:"
)

print(
    "Binary Particle Swarm Optimization"
)


print(
    "\nPSO best fitness:",
    round(
        global_best_fitness,
        6
    )
)


print(
    "Original features:",
    NUMBER_OF_FEATURES
)


print(
    "Selected features:",
    len(
        selected_features
    )
)


print(
    f"\nAccuracy : "
    f"{accuracy * 100:.2f}%"
)


print(
    f"Precision: "
    f"{precision * 100:.2f}%"
)


print(
    f"Recall   : "
    f"{recall * 100:.2f}%"
)


print(
    f"F1 Score : "
    f"{f1 * 100:.2f}%"
)


print(
    f"ROC-AUC  : "
    f"{roc_auc * 100:.2f}%"
)


print(
    "\nTotal transactions:",
    TOTAL_RECORDS
)


print(
    "Clean transactions:",
    CLEAN_RECORDS
)


print(
    "Anomalous transactions:",
    ANOMALY_RECORDS
)


print(
    "\nOutput directory:"
)

print(
    PROJECT_DIR
)


print(
    "\nAll PSO results have been saved "
    "using the pso_ prefix."
)


print("=" * 80)
```
