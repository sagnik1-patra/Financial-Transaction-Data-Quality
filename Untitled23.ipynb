```python
import os
import json
import random
import warnings
from pathlib import Path

import joblib
import numpy as np
import pandas as pd
import yaml

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    classification_report,
    roc_auc_score
)

import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Dropout, BatchNormalization
from tensorflow.keras.callbacks import EarlyStopping, ReduceLROnPlateau


# ================================================================
# 2. SETTINGS
# ================================================================

warnings.filterwarnings("ignore")

SEED = 42

random.seed(SEED)
np.random.seed(SEED)
tf.random.set_seed(SEED)


# ================================================================
# 3. PATHS
# ================================================================

PROJECT_DIR = Path(
    r"C:\Users\sagni\Downloads\Financial Transaction Data Quality"
)

CSV_PATH = PROJECT_DIR / "dirty_financial_transactions.csv"


# Output files

H5_PATH = PROJECT_DIR / "financial_data_quality_model.h5"

PKL_PATH = PROJECT_DIR / "financial_data_quality_pipeline.pkl"

YAML_PATH = PROJECT_DIR / "financial_data_quality_config.yaml"

JSON_PATH = PROJECT_DIR / "financial_data_quality_results.json"

LABELED_CSV_PATH = (
    PROJECT_DIR / "financial_data_quality_labeled.csv"
)

PREDICTION_CSV_PATH = (
    PROJECT_DIR / "financial_data_quality_predictions.csv"
)


# Create folder if it does not exist

PROJECT_DIR.mkdir(
    parents=True,
    exist_ok=True
)


# ================================================================
# 4. LOAD DATASET
# ================================================================

print("\n" + "=" * 70)
print("FINANCIAL TRANSACTION DATA QUALITY PREDICTION")
print("=" * 70)

print("\nLoading dataset...")

if not CSV_PATH.exists():
    raise FileNotFoundError(
        f"\nDataset not found:\n{CSV_PATH}"
    )


df = pd.read_csv(
    CSV_PATH,
    low_memory=False
)


print("\nDataset loaded successfully.")

print(
    "\nDataset shape:",
    df.shape
)

print(
    "\nNumber of rows:",
    len(df)
)

print(
    "\nNumber of columns:",
    len(df.columns)
)


print("\nColumns:")

for column in df.columns:
    print(" -", column)


print("\nFirst 5 rows:")

print(
    df.head()
)


# ================================================================
# 5. REQUIRED COLUMNS
# ================================================================

required_columns = [

    "Transaction_ID",
    "Transaction_Date",
    "Customer_ID",
    "Product_Name",
    "Quantity",
    "Price",
    "Payment_Method",
    "Transaction_Status"

]


missing_columns = [

    col
    for col in required_columns
    if col not in df.columns

]


if missing_columns:

    raise ValueError(
        f"Missing required columns: "
        f"{missing_columns}"
    )


# ================================================================
# 6. KEEP ORIGINAL DATA
# ================================================================

original_df = df.copy()


# ================================================================
# 7. HELPER FUNCTIONS
# ================================================================


def clean_string(value):
    """
    Convert a value into a safely stripped string.
    """

    if pd.isna(value):
        return ""

    return str(value).strip()



def normalize_text(value):
    """
    Lowercase and remove unnecessary spaces.
    """

    value = clean_string(value)

    return value.lower().strip()



def clean_price(value):
    """
    Convert price into numeric form.

    Examples:
        "$420.21" -> 420.21
        "500"     -> 500.0
        NaN       -> NaN
    """

    if pd.isna(value):
        return np.nan

    value = str(value)

    value = value.replace(
        "$",
        ""
    )

    value = value.replace(
        ",",
        ""
    )

    value = value.strip()

    try:

        return float(value)

    except:

        return np.nan



def valid_transaction_id(value):
    """
    A valid transaction ID should look like:

    T0001
    T0123
    etc.
    """

    value = clean_string(value)

    if value == "":
        return False

    if not value.startswith("T"):
        return False

    number_part = value[1:]

    if not number_part.isdigit():
        return False

    return True



def valid_customer_id(value):
    """
    Check that customer ID exists.

    We avoid making the format too restrictive
    because different datasets may use
    different customer-ID structures.
    """

    value = clean_string(value)

    if value == "":
        return False

    return True



# ================================================================
# 8. DEFINE CANONICAL VALUES
# ================================================================


VALID_PRODUCTS = {

    "Laptop",
    "Tablet",
    "Smartphone",
    "Headphones",
    "Coffee Machine"

}


VALID_PAYMENT_METHODS = {

    "Credit Card",
    "PayPal",
    "Cash"

}


VALID_TRANSACTION_STATUS = {

    "Completed",
    "Pending",
    "Failed"

}


# ================================================================
# 9. CLEAN / STANDARDIZE CATEGORICAL DATA
# ================================================================


def standardize_payment_method(value):

    text = normalize_text(value)

    text_no_space = text.replace(
        " ",
        ""
    )

    if text_no_space == "creditcard":

        return "Credit Card"

    elif text_no_space == "paypal":

        return "PayPal"

    elif text == "cash":

        return "Cash"

    return "Unknown"



def standardize_status(value):

    text = normalize_text(value)

    if text in [
        "completed",
        "complete"
    ]:

        return "Completed"

    elif text == "pending":

        return "Pending"

    elif text == "failed":

        return "Failed"

    return "Unknown"



def standardize_product(value):

    text = normalize_text(value)

    product_mapping = {

        "laptop": "Laptop",

        "tablet": "Tablet",

        "smartphone": "Smartphone",

        "headphones": "Headphones",

        "coffee machine": "Coffee Machine"

    }

    return product_mapping.get(
        text,
        "Unknown"
    )


# ================================================================
# 10. BASIC CLEANED COLUMNS
# ================================================================


df["Price_Numeric"] = (
    df["Price"]
    .apply(clean_price)
)


df["Parsed_Date"] = pd.to_datetime(
    df["Transaction_Date"],
    errors="coerce"
)


df["Payment_Standardized"] = (
    df["Payment_Method"]
    .apply(standardize_payment_method)
)


df["Status_Standardized"] = (
    df["Transaction_Status"]
    .apply(standardize_status)
)


df["Product_Standardized"] = (
    df["Product_Name"]
    .apply(standardize_product)
)


# ================================================================
# 11. CREATE DATA QUALITY RULES
# ================================================================
#
# Each rule returns:
#
# 0 = valid
# 1 = wrong / anomalous
#
# ================================================================


# ------------------------------------------------
# Rule 1: Missing Transaction ID
# ------------------------------------------------

df["Error_Missing_Transaction_ID"] = (
    df["Transaction_ID"]
    .isna()
    .astype(int)
)


# ------------------------------------------------
# Rule 2: Invalid Transaction ID
# ------------------------------------------------

df["Error_Invalid_Transaction_ID"] = (

    ~df["Transaction_ID"]
    .apply(valid_transaction_id)

).astype(int)


# ------------------------------------------------
# Rule 3: Missing Customer ID
# ------------------------------------------------

df["Error_Missing_Customer_ID"] = (
    df["Customer_ID"]
    .isna()
    .astype(int)
)


# ------------------------------------------------
# Rule 4: Invalid Customer ID
# ------------------------------------------------

df["Error_Invalid_Customer_ID"] = (

    ~df["Customer_ID"]
    .apply(valid_customer_id)

).astype(int)


# ------------------------------------------------
# Rule 5: Missing Date
# ------------------------------------------------

df["Error_Missing_Date"] = (
    df["Transaction_Date"]
    .isna()
    .astype(int)
)


# ------------------------------------------------
# Rule 6: Invalid Date
# ------------------------------------------------

df["Error_Invalid_Date"] = (

    (
        df["Transaction_Date"].notna()
    )
    &
    (
        df["Parsed_Date"].isna()
    )

).astype(int)


# ------------------------------------------------
# Rule 7: Missing Product
# ------------------------------------------------

df["Error_Missing_Product"] = (
    df["Product_Name"]
    .isna()
    .astype(int)
)


# ------------------------------------------------
# Rule 8: Invalid / Corrupted Product
# ------------------------------------------------

df["Error_Invalid_Product"] = (

    (
        df["Product_Name"].notna()
    )
    &
    (
        df["Product_Standardized"]
        == "Unknown"
    )

).astype(int)


# ------------------------------------------------
# Rule 9: Missing Quantity
# ------------------------------------------------

df["Error_Missing_Quantity"] = (
    df["Quantity"]
    .isna()
    .astype(int)
)


# ------------------------------------------------
# Rule 10: Invalid Quantity
# ------------------------------------------------

df["Error_Invalid_Quantity"] = (

    (
        df["Quantity"].notna()
    )
    &
    (
        df["Quantity"] <= 0
    )

).astype(int)


# ------------------------------------------------
# Rule 11: Missing Price
# ------------------------------------------------

df["Error_Missing_Price"] = (
    df["Price"]
    .isna()
    .astype(int)
)


# ------------------------------------------------
# Rule 12: Invalid Price Format
# ------------------------------------------------

df["Error_Invalid_Price_Format"] = (

    (
        df["Price"].notna()
    )
    &
    (
        df["Price_Numeric"].isna()
    )

).astype(int)


# ------------------------------------------------
# Rule 13: Negative / Zero Price
# ------------------------------------------------

df["Error_Invalid_Price_Value"] = (

    (
        df["Price_Numeric"].notna()
    )
    &
    (
        df["Price_Numeric"] <= 0
    )

).astype(int)


# ------------------------------------------------
# Rule 14: Missing Payment Method
# ------------------------------------------------

df["Error_Missing_Payment_Method"] = (
    df["Payment_Method"]
    .isna()
    .astype(int)
)


# ------------------------------------------------
# Rule 15: Non-standard Payment Method
# ------------------------------------------------
#
# "creditcard" and "pay pal" can be normalized,
# but they are still considered dirty in the
# original dataset.
#
# ------------------------------------------------

payment_original = (
    df["Payment_Method"]
    .fillna("")
    .astype(str)
    .str.strip()
)


df["Error_Nonstandard_Payment"] = (

    (
        df["Payment_Method"].notna()
    )
    &
    (
        ~payment_original.isin(
            VALID_PAYMENT_METHODS
        )
    )

).astype(int)


# ------------------------------------------------
# Rule 16: Missing Transaction Status
# ------------------------------------------------

df["Error_Missing_Status"] = (
    df["Transaction_Status"]
    .isna()
    .astype(int)
)


# ------------------------------------------------
# Rule 17: Non-standard Status
# ------------------------------------------------

status_original = (
    df["Transaction_Status"]
    .fillna("")
    .astype(str)
    .str.strip()
)


df["Error_Nonstandard_Status"] = (

    (
        df["Transaction_Status"]
        .notna()
    )
    &
    (
        ~status_original.isin(
            VALID_TRANSACTION_STATUS
        )
    )

).astype(int)


# ------------------------------------------------
# Rule 18: Duplicate Transaction ID
# ------------------------------------------------

df["Error_Duplicate_Transaction_ID"] = (

    df["Transaction_ID"]
    .duplicated(
        keep=False
    )

    &

    df["Transaction_ID"]
    .notna()

).astype(int)


# ================================================================
# 12. LIST ALL ERROR COLUMNS
# ================================================================

ERROR_COLUMNS = [

    "Error_Missing_Transaction_ID",

    "Error_Invalid_Transaction_ID",

    "Error_Missing_Customer_ID",

    "Error_Invalid_Customer_ID",

    "Error_Missing_Date",

    "Error_Invalid_Date",

    "Error_Missing_Product",

    "Error_Invalid_Product",

    "Error_Missing_Quantity",

    "Error_Invalid_Quantity",

    "Error_Missing_Price",

    "Error_Invalid_Price_Format",

    "Error_Invalid_Price_Value",

    "Error_Missing_Payment_Method",

    "Error_Nonstandard_Payment",

    "Error_Missing_Status",

    "Error_Nonstandard_Status",

    "Error_Duplicate_Transaction_ID"

]


# ================================================================
# 13. CREATE TARGET VARIABLE
# ================================================================
#
# Is_Anomaly:
#
# 0 = clean transaction
# 1 = dirty / anomalous transaction
#
# ================================================================


df["Error_Count"] = (
    df[ERROR_COLUMNS]
    .sum(axis=1)
)


df["Is_Anomaly"] = (

    df["Error_Count"] > 0

).astype(int)


# ================================================================
# 14. CREATE ERROR DESCRIPTION
# ================================================================


ERROR_NAMES = {

    "Error_Missing_Transaction_ID":
        "Missing Transaction ID",

    "Error_Invalid_Transaction_ID":
        "Invalid Transaction ID",

    "Error_Missing_Customer_ID":
        "Missing Customer ID",

    "Error_Invalid_Customer_ID":
        "Invalid Customer ID",

    "Error_Missing_Date":
        "Missing Date",

    "Error_Invalid_Date":
        "Invalid Date",

    "Error_Missing_Product":
        "Missing Product",

    "Error_Invalid_Product":
        "Invalid Product",

    "Error_Missing_Quantity":
        "Missing Quantity",

    "Error_Invalid_Quantity":
        "Invalid Quantity",

    "Error_Missing_Price":
        "Missing Price",

    "Error_Invalid_Price_Format":
        "Invalid Price Format",

    "Error_Invalid_Price_Value":
        "Invalid Price Value",

    "Error_Missing_Payment_Method":
        "Missing Payment Method",

    "Error_Nonstandard_Payment":
        "Nonstandard Payment Method",

    "Error_Missing_Status":
        "Missing Transaction Status",

    "Error_Nonstandard_Status":
        "Nonstandard Transaction Status",

    "Error_Duplicate_Transaction_ID":
        "Duplicate Transaction ID"

}


def get_error_description(row):

    errors = []

    for column in ERROR_COLUMNS:

        if row[column] == 1:

            errors.append(
                ERROR_NAMES[column]
            )

    if len(errors) == 0:

        return "Clean"

    return " | ".join(errors)


df["Error_Type"] = df.apply(
    get_error_description,
    axis=1
)


# ================================================================
# 15. DISPLAY DATA QUALITY SUMMARY
# ================================================================

print("\n" + "=" * 70)
print("DATA QUALITY ANALYSIS")
print("=" * 70)


clean_count = int(
    (df["Is_Anomaly"] == 0).sum()
)


anomaly_count = int(
    (df["Is_Anomaly"] == 1).sum()
)


total_records = len(df)


clean_percentage = (
    clean_count
    / total_records
    * 100
)


anomaly_percentage = (
    anomaly_count
    / total_records
    * 100
)


print(
    f"\nTotal records: "
    f"{total_records}"
)

print(
    f"Clean records: "
    f"{clean_count}"
)

print(
    f"Anomalous records: "
    f"{anomaly_count}"
)

print(
    f"Clean percentage: "
    f"{clean_percentage:.2f}%"
)

print(
    f"Anomaly percentage: "
    f"{anomaly_percentage:.2f}%"
)


print(
    "\nError distribution:"
)


for column in ERROR_COLUMNS:

    count = int(
        df[column].sum()
    )

    print(
        f"{ERROR_NAMES[column]:35s}: "
        f"{count}"
    )


# ================================================================
# 16. FEATURE ENGINEERING FOR ML
# ================================================================
#
# IMPORTANT:
#
# We do NOT directly give the neural network
# the Is_Anomaly target or Error_Count.
#
# Instead, features are generated from
# properties of each raw transaction.
#
# ================================================================


features = pd.DataFrame(
    index=df.index
)


# ------------------------------------------------
# Transaction ID features
# ------------------------------------------------

features[
    "transaction_id_missing"
] = (
    df["Transaction_ID"]
    .isna()
    .astype(int)
)


features[
    "transaction_id_length"
] = (
    df["Transaction_ID"]
    .fillna("")
    .astype(str)
    .str.len()
)


features[
    "transaction_id_valid_format"
] = (
    df["Transaction_ID"]
    .apply(valid_transaction_id)
    .astype(int)
)


features[
    "transaction_id_duplicate"
] = (
    df["Transaction_ID"]
    .duplicated(
        keep=False
    )
    .astype(int)
)


# ------------------------------------------------
# Customer ID features
# ------------------------------------------------

features[
    "customer_id_missing"
] = (
    df["Customer_ID"]
    .isna()
    .astype(int)
)


features[
    "customer_id_length"
] = (
    df["Customer_ID"]
    .fillna("")
    .astype(str)
    .str.len()
)


# ------------------------------------------------
# Date features
# ------------------------------------------------

features[
    "date_missing"
] = (
    df["Transaction_Date"]
    .isna()
    .astype(int)
)


features[
    "date_valid"
] = (
    df["Parsed_Date"]
    .notna()
    .astype(int)
)


features[
    "date_year"
] = (
    df["Parsed_Date"]
    .dt.year
    .fillna(0)
)


features[
    "date_month"
] = (
    df["Parsed_Date"]
    .dt.month
    .fillna(0)
)


features[
    "date_day"
] = (
    df["Parsed_Date"]
    .dt.day
    .fillna(0)
)


# ------------------------------------------------
# Product features
# ------------------------------------------------

features[
    "product_missing"
] = (
    df["Product_Name"]
    .isna()
    .astype(int)
)


features[
    "product_name_length"
] = (
    df["Product_Name"]
    .fillna("")
    .astype(str)
    .str.strip()
    .str.len()
)


features[
    "product_standard"
] = (
    df["Product_Name"]
    .fillna("")
    .astype(str)
    .str.strip()
    .isin(
        VALID_PRODUCTS
    )
    .astype(int)
)


# ------------------------------------------------
# Quantity features
# ------------------------------------------------

quantity_numeric = pd.to_numeric(
    df["Quantity"],
    errors="coerce"
)


features[
    "quantity_missing"
] = (
    quantity_numeric
    .isna()
    .astype(int)
)


features[
    "quantity_value"
] = (
    quantity_numeric
    .fillna(0)
)


features[
    "quantity_positive"
] = (
    quantity_numeric
    .gt(0)
    .fillna(False)
    .astype(int)
)


features[
    "quantity_negative"
] = (
    quantity_numeric
    .lt(0)
    .fillna(False)
    .astype(int)
)


# ------------------------------------------------
# Price features
# ------------------------------------------------

features[
    "price_missing"
] = (
    df["Price"]
    .isna()
    .astype(int)
)


features[
    "price_numeric"
] = (
    df["Price_Numeric"]
    .fillna(0)
)


features[
    "price_positive"
] = (
    df["Price_Numeric"]
    .gt(0)
    .fillna(False)
    .astype(int)
)


features[
    "price_negative"
] = (
    df["Price_Numeric"]
    .lt(0)
    .fillna(False)
    .astype(int)
)


features[
    "price_contains_dollar"
] = (
    df["Price"]
    .fillna("")
    .astype(str)
    .str.contains(
        r"\$",
        regex=True
    )
    .astype(int)
)


features[
    "price_text_length"
] = (
    df["Price"]
    .fillna("")
    .astype(str)
    .str.len()
)


# ------------------------------------------------
# Payment Method features
# ------------------------------------------------

features[
    "payment_missing"
] = (
    df["Payment_Method"]
    .isna()
    .astype(int)
)


features[
    "payment_standard"
] = (
    payment_original
    .isin(
        VALID_PAYMENT_METHODS
    )
    .astype(int)
)


features[
    "payment_length"
] = (
    payment_original
    .str.len()
)


# ------------------------------------------------
# Status features
# ------------------------------------------------

features[
    "status_missing"
] = (
    df["Transaction_Status"]
    .isna()
    .astype(int)
)


features[
    "status_standard"
] = (
    status_original
    .isin(
        VALID_TRANSACTION_STATUS
    )
    .astype(int)
)


features[
    "status_length"
] = (
    status_original
    .str.len()
)


# ================================================================
# 17. HANDLE INFINITE / MISSING VALUES
# ================================================================

features = features.replace(
    [np.inf, -np.inf],
    np.nan
)


features = features.fillna(0)


# Convert everything to float

features = features.astype(
    np.float32
)


print("\n" + "=" * 70)
print("ML FEATURE ENGINEERING")
print("=" * 70)


print(
    "\nNumber of engineered features:",
    features.shape[1]
)


print(
    "\nFeature names:"
)


for feature in features.columns:

    print(
        " -",
        feature
    )


# ================================================================
# 18. TARGET VARIABLE
# ================================================================

X = features.copy()


y = (
    df["Is_Anomaly"]
    .astype(np.int32)
)


# ================================================================
# 19. TRAIN / TEST SPLIT
# ================================================================


X_train, X_test, y_train, y_test = (
    train_test_split(

        X,
        y,

        test_size=0.20,

        random_state=SEED,

        stratify=y

    )
)


print("\n" + "=" * 70)
print("TRAIN TEST SPLIT")
print("=" * 70)


print(
    "\nTraining samples:",
    len(X_train)
)


print(
    "Testing samples:",
    len(X_test)
)


# ================================================================
# 20. FEATURE SCALING
# ================================================================

scaler = StandardScaler()


X_train_scaled = scaler.fit_transform(
    X_train
)


X_test_scaled = scaler.transform(
    X_test
)


# ================================================================
# 21. CALCULATE CLASS WEIGHTS
# ================================================================

class_counts = np.bincount(
    y_train
)


class_weight = {}


if len(class_counts) == 2:

    total = len(y_train)

    class_weight[0] = (
        total
        /
        (
            2.0
            *
            class_counts[0]
        )
    )

    class_weight[1] = (
        total
        /
        (
            2.0
            *
            class_counts[1]
        )
    )

else:

    class_weight = None


print(
    "\nClass weights:",
    class_weight
)


# ================================================================
# 22. BUILD NEURAL NETWORK MODEL
# ================================================================


model = Sequential(

    [

        tf.keras.Input(
            shape=(
                X_train_scaled.shape[1],
            )
        ),

        Dense(
            128,
            activation="relu"
        ),

        BatchNormalization(),

        Dropout(
            0.30
        ),

        Dense(
            64,
            activation="relu"
        ),

        BatchNormalization(),

        Dropout(
            0.25
        ),

        Dense(
            32,
            activation="relu"
        ),

        Dropout(
            0.20
        ),

        Dense(
            16,
            activation="relu"
        ),

        Dense(
            1,
            activation="sigmoid"
        )

    ]

)


# ================================================================
# 23. COMPILE MODEL
# ================================================================


model.compile(

    optimizer=tf.keras.optimizers.Adam(
        learning_rate=0.001
    ),

    loss="binary_crossentropy",

    metrics=[
        "accuracy",
        tf.keras.metrics.Precision(
            name="precision"
        ),
        tf.keras.metrics.Recall(
            name="recall"
        ),
        tf.keras.metrics.AUC(
            name="auc"
        )
    ]

)


print("\n" + "=" * 70)
print("NEURAL NETWORK")
print("=" * 70)


model.summary()


# ================================================================
# 24. CALLBACKS
# ================================================================


early_stopping = EarlyStopping(

    monitor="val_loss",

    patience=7,

    restore_best_weights=True,

    verbose=1

)


reduce_lr = ReduceLROnPlateau(

    monitor="val_loss",

    factor=0.5,

    patience=3,

    min_lr=0.000001,

    verbose=1

)


# ================================================================
# 25. TRAIN MODEL
# ================================================================


print("\n" + "=" * 70)
print("TRAINING MODEL")
print("=" * 70)


history = model.fit(

    X_train_scaled,
    y_train,

    validation_split=0.20,

    epochs=50,

    batch_size=256,

    callbacks=[
        early_stopping,
        reduce_lr
    ],

    class_weight=class_weight,

    verbose=1

)


# ================================================================
# 26. MODEL PREDICTION
# ================================================================


print("\nPredicting test data...")


prediction_probability = (

    model.predict(
        X_test_scaled,
        verbose=0
    )

    .flatten()

)


y_pred = (

    prediction_probability
    >= 0.50

).astype(int)


# ================================================================
# 27. EVALUATION
# ================================================================


accuracy = accuracy_score(
    y_test,
    y_pred
)


precision = precision_score(
    y_test,
    y_pred,
    zero_division=0
)


recall = recall_score(
    y_test,
    y_pred,
    zero_division=0
)


f1 = f1_score(
    y_test,
    y_pred,
    zero_division=0
)


try:

    roc_auc = roc_auc_score(
        y_test,
        prediction_probability
    )

except:

    roc_auc = 0.0


cm = confusion_matrix(
    y_test,
    y_pred
)


report = classification_report(

    y_test,
    y_pred,

    target_names=[
        "Clean",
        "Anomaly"
    ],

    output_dict=True,

    zero_division=0

)


print("\n" + "=" * 70)
print("MODEL RESULTS")
print("=" * 70)


print(
    f"\nAccuracy : "
    f"{accuracy:.4f}"
)


print(
    f"Precision: "
    f"{precision:.4f}"
)


print(
    f"Recall   : "
    f"{recall:.4f}"
)


print(
    f"F1 Score : "
    f"{f1:.4f}"
)


print(
    f"ROC-AUC  : "
    f"{roc_auc:.4f}"
)


print(
    "\nConfusion Matrix:"
)

print(
    cm
)


print(
    "\nClassification Report:"
)


print(

    classification_report(

        y_test,
        y_pred,

        target_names=[
            "Clean",
            "Anomaly"
        ],

        zero_division=0

    )

)


# ================================================================
# 28. SAVE H5 MODEL
# ================================================================


print("\nSaving H5 model...")


model.save(
    H5_PATH
)


print(
    "Saved:",
    H5_PATH
)


# ================================================================
# 29. SAVE PKL PIPELINE
# ================================================================
#
# Stores everything required for preprocessing
# future transactions.
#
# ================================================================


pipeline_data = {

    "scaler":
        scaler,

    "feature_names":
        list(X.columns),

    "threshold":
        0.50,

    "valid_products":
        list(VALID_PRODUCTS),

    "valid_payment_methods":
        list(VALID_PAYMENT_METHODS),

    "valid_transaction_status":
        list(VALID_TRANSACTION_STATUS),

    "error_columns":
        ERROR_COLUMNS,

    "error_names":
        ERROR_NAMES,

    "random_seed":
        SEED

}


joblib.dump(
    pipeline_data,
    PKL_PATH
)


print(
    "Saved:",
    PKL_PATH
)


# ================================================================
# 30. SAVE YAML CONFIGURATION
# ================================================================


yaml_config = {

    "project": {

        "name":
            "Financial Transaction Data Quality Prediction",

        "description":
            "Machine learning system for detecting dirty, "
            "incorrect, incomplete and anomalous financial "
            "transaction records.",

        "dataset":
            str(CSV_PATH),

        "output_directory":
            str(PROJECT_DIR)

    },

    "dataset": {

        "rows":
            int(df.shape[0]),

        "columns":
            int(original_df.shape[1]),

        "column_names":
            list(original_df.columns)

    },

    "target": {

        "column":
            "Is_Anomaly",

        "clean_class":
            0,

        "anomaly_class":
            1

    },

    "model": {

        "type":
            "Artificial Neural Network",

        "framework":
            "TensorFlow / Keras",

        "input_features":
            int(X.shape[1]),

        "hidden_layers": [
            128,
            64,
            32,
            16
        ],

        "output_units":
            1,

        "output_activation":
            "sigmoid",

        "optimizer":
            "Adam",

        "learning_rate":
            0.001,

        "loss":
            "binary_crossentropy",

        "epochs":
            50,

        "batch_size":
            256,

        "prediction_threshold":
            0.50

    },

    "training": {

        "train_percentage":
            80,

        "test_percentage":
            20,

        "random_state":
            SEED

    },

    "data_quality_rules": [

        ERROR_NAMES[
            column
        ]

        for column
        in ERROR_COLUMNS

    ],

    "output_files": {

        "h5_model":
            str(H5_PATH),

        "pkl_pipeline":
            str(PKL_PATH),

        "yaml_config":
            str(YAML_PATH),

        "json_results":
            str(JSON_PATH),

        "labeled_csv":
            str(LABELED_CSV_PATH),

        "predictions_csv":
            str(PREDICTION_CSV_PATH)

    }

}


with open(
    YAML_PATH,
    "w",
    encoding="utf-8"
) as yaml_file:

    yaml.safe_dump(

        yaml_config,

        yaml_file,

        sort_keys=False,

        allow_unicode=True

    )


print(
    "Saved:",
    YAML_PATH
)


# ================================================================
# 31. SAVE JSON RESULTS
# ================================================================


error_distribution = {}


for column in ERROR_COLUMNS:

    error_distribution[
        ERROR_NAMES[column]
    ] = int(
        df[column].sum()
    )


training_history = {}


for key, values in history.history.items():

    training_history[key] = [

        float(value)
        for value in values

    ]


json_results = {

    "project_name":
        "Financial Transaction Data Quality Prediction",

    "total_records":
        int(total_records),

    "clean_records":
        int(clean_count),

    "anomalous_records":
        int(anomaly_count),

    "clean_percentage":
        float(clean_percentage),

    "anomaly_percentage":
        float(anomaly_percentage),

    "training_records":
        int(len(X_train)),

    "testing_records":
        int(len(X_test)),

    "number_of_features":
        int(X.shape[1]),

    "feature_names":
        list(X.columns),

    "metrics": {

        "accuracy":
            float(accuracy),

        "precision":
            float(precision),

        "recall":
            float(recall),

        "f1_score":
            float(f1),

        "roc_auc":
            float(roc_auc)

    },

    "confusion_matrix":
        cm.tolist(),

    "classification_report":
        report,

    "error_distribution":
        error_distribution,

    "training_history":
        training_history

}


with open(
    JSON_PATH,
    "w",
    encoding="utf-8"
) as json_file:

    json.dump(

        json_results,

        json_file,

        indent=4,

        ensure_ascii=False

    )


print(
    "Saved:",
    JSON_PATH
)


# ================================================================
# 32. SAVE LABELED DATASET
# ================================================================


labeled_output = original_df.copy()


for column in ERROR_COLUMNS:

    labeled_output[column] = (
        df[column]
    )


labeled_output[
    "Error_Count"
] = df["Error_Count"]


labeled_output[
    "Error_Type"
] = df["Error_Type"]


labeled_output[
    "Is_Anomaly"
] = df["Is_Anomaly"]


labeled_output.to_csv(

    LABELED_CSV_PATH,

    index=False

)


print(
    "Saved:",
    LABELED_CSV_PATH
)


# ================================================================
# 33. SAVE TEST PREDICTIONS
# ================================================================


prediction_output = (

    original_df
    .loc[X_test.index]
    .copy()

)


prediction_output[
    "Actual_Label"
] = (
    y_test.values
)


prediction_output[
    "Actual_Class"
] = np.where(

    y_test.values == 1,

    "Anomaly",

    "Clean"

)


prediction_output[
    "Prediction_Probability"
] = (
    prediction_probability
)


prediction_output[
    "Predicted_Label"
] = (
    y_pred
)


prediction_output[
    "Predicted_Class"
] = np.where(

    y_pred == 1,

    "Anomaly",

    "Clean"

)


prediction_output[
    "Prediction_Correct"
] = (

    prediction_output[
        "Actual_Label"
    ]

    ==

    prediction_output[
        "Predicted_Label"
    ]

)


prediction_output.to_csv(

    PREDICTION_CSV_PATH,

    index=False

)


print(
    "Saved:",
    PREDICTION_CSV_PATH
)


# ================================================================
# 34. VERIFY SAVED MODEL
# ================================================================


print("\n" + "=" * 70)
print("VERIFYING SAVED H5 MODEL")
print("=" * 70)


loaded_model = (
    tf.keras.models.load_model(
        H5_PATH
    )
)


sample_predictions = (

    loaded_model.predict(

        X_test_scaled[:5],

        verbose=0

    )

    .flatten()

)


print(
    "\nFirst 5 probabilities:"
)


for i, probability in enumerate(
    sample_predictions,
    start=1
):

    predicted_class = (

        "ANOMALY"

        if probability >= 0.50

        else "CLEAN"

    )

    print(

        f"Sample {i}: "
        f"{probability:.4f} "
        f"-> {predicted_class}"

    )


# ================================================================
# 35. VERIFY PKL FILE
# ================================================================


loaded_pipeline = joblib.load(
    PKL_PATH
)


print(
    "\nPKL pipeline loaded successfully."
)


print(
    "Number of stored features:",
    len(
        loaded_pipeline[
            "feature_names"
        ]
    )
)


# ================================================================
# 36. FINAL SUMMARY
# ================================================================


print("\n" + "=" * 70)

print(
    "PROJECT COMPLETED SUCCESSFULLY"
)

print("=" * 70)


print(
    "\nDataset:"
)

print(
    CSV_PATH
)


print(
    "\nOutput directory:"
)

print(
    PROJECT_DIR
)


print(
    "\nGenerated Files:"
)


print(
    "\n1.",
    H5_PATH.name
)


print(
    "2.",
    PKL_PATH.name
)


print(
    "3.",
    YAML_PATH.name
)


print(
    "4.",
    JSON_PATH.name
)


print(
    "5.",
    LABELED_CSV_PATH.name
)


print(
    "6.",
    PREDICTION_CSV_PATH.name
)


print(
    "\nFinal Model Accuracy:"
)


print(
    f"{accuracy * 100:.2f}%"
)


print(
    "\nF1 Score:"
)


print(
    f"{f1 * 100:.2f}%"
)


print(
    "\nROC-AUC:"
)


print(
    f"{roc_auc:.4f}"
)


print(
    "\nClean Transactions:",
    clean_count
)


print(
    "Anomalous Transactions:",
    anomaly_count
)


print(
    "\nAll required H5, PKL, YAML and JSON "
    "files have been generated."
)


print("\n" + "=" * 70)
```
