```python
import os
import json
import random
import warnings
from pathlib import Path

import numpy as np
import pandas as pd

import matplotlib.pyplot as plt

import joblib
import yaml

from sklearn.model_selection import (
    train_test_split,
    StratifiedShuffleSplit
)

from sklearn.preprocessing import StandardScaler

from sklearn.linear_model import LogisticRegression

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    confusion_matrix,
    classification_report
)

import tensorflow as tf

from tensorflow.keras.models import Sequential

from tensorflow.keras.layers import (
    Dense,
    Dropout,
    BatchNormalization
)

from tensorflow.keras.callbacks import (
    EarlyStopping,
    ReduceLROnPlateau
)


# =====================================================================
# 2. SETTINGS
# =====================================================================

warnings.filterwarnings("ignore")

SEED = 42

random.seed(SEED)
np.random.seed(SEED)
tf.random.set_seed(SEED)


# =====================================================================
# 3. PATHS
# =====================================================================

PROJECT_DIR = Path(
    r"C:\Users\sagni\Downloads\Financial Transaction Data Quality"
)

CSV_PATH = (
    PROJECT_DIR
    / "dirty_financial_transactions.csv"
)


PROJECT_DIR.mkdir(
    parents=True,
    exist_ok=True
)


# =====================================================================
# 4. AIS OUTPUT PATHS
# =====================================================================

H5_PATH = (
    PROJECT_DIR
    / "ais_financial_data_quality_model.h5"
)

PKL_PATH = (
    PROJECT_DIR
    / "ais_financial_data_quality_pipeline.pkl"
)

YAML_PATH = (
    PROJECT_DIR
    / "ais_financial_data_quality_config.yaml"
)

JSON_PATH = (
    PROJECT_DIR
    / "ais_financial_data_quality_results.json"
)


ACCURACY_GRAPH_PATH = (
    PROJECT_DIR
    / "ais_accuracy_graph.png"
)

HEATMAP_PATH = (
    PROJECT_DIR
    / "ais_heatmap.png"
)

COMPARISON_GRAPH_PATH = (
    PROJECT_DIR
    / "ais_comparison_graph.png"
)

RESULT_CSV_PATH = (
    PROJECT_DIR
    / "ais_result.csv"
)

RESULT_GRAPH_PATH = (
    PROJECT_DIR
    / "ais_result_graph.png"
)

PREDICTION_CSV_PATH = (
    PROJECT_DIR
    / "ais_prediction.csv"
)

PREDICTION_GRAPH_PATH = (
    PROJECT_DIR
    / "ais_prediction_graph.png"
)

FEATURE_SELECTION_CSV_PATH = (
    PROJECT_DIR
    / "ais_feature_selection.csv"
)

FEATURE_SELECTION_GRAPH_PATH = (
    PROJECT_DIR
    / "ais_feature_selection_graph.png"
)

LABELED_CSV_PATH = (
    PROJECT_DIR
    / "ais_labeled_dataset.csv"
)

AIS_HISTORY_CSV_PATH = (
    PROJECT_DIR
    / "ais_optimization_history.csv"
)

AIS_HISTORY_GRAPH_PATH = (
    PROJECT_DIR
    / "ais_optimization_history_graph.png"
)


# =====================================================================
# 5. LOAD DATASET
# =====================================================================

print("\n" + "=" * 80)
print("AIS-BASED FINANCIAL TRANSACTION DATA QUALITY PREDICTION")
print("=" * 80)


if not CSV_PATH.exists():

    raise FileNotFoundError(
        f"Dataset not found:\n{CSV_PATH}"
    )


df = pd.read_csv(
    CSV_PATH,
    low_memory=False
)


original_df = df.copy()


print(
    "\nDataset loaded:",
    df.shape
)


# =====================================================================
# 6. VALIDATE COLUMNS
# =====================================================================

required_columns = [

    "Transaction_ID",
    "Transaction_Date",
    "Customer_ID",
    "Product_Name",
    "Quantity",
    "Price",
    "Payment_Method",
    "Transaction_Status"

]


missing_columns = [

    col
    for col in required_columns
    if col not in df.columns

]


if missing_columns:

    raise ValueError(
        f"Missing columns: {missing_columns}"
    )


# =====================================================================
# 7. HELPER FUNCTIONS
# =====================================================================


def safe_string(value):

    if pd.isna(value):

        return ""

    return str(value).strip()



def clean_price(value):

    if pd.isna(value):

        return np.nan

    value = str(value)

    value = value.replace(
        "$",
        ""
    )

    value = value.replace(
        ",",
        ""
    )

    value = value.strip()

    try:

        return float(value)

    except:

        return np.nan



def valid_transaction_id(value):

    value = safe_string(
        value
    )

    if value == "":

        return False

    if not value.startswith("T"):

        return False

    return value[1:].isdigit()



def valid_customer_id(value):

    value = safe_string(
        value
    )

    return value != ""


# =====================================================================
# 8. VALID CATEGORY VALUES
# =====================================================================

VALID_PRODUCTS = {

    "Laptop",
    "Tablet",
    "Smartphone",
    "Headphones",
    "Coffee Machine"

}


VALID_PAYMENT_METHODS = {

    "Credit Card",
    "PayPal",
    "Cash"

}


VALID_STATUSES = {

    "Completed",
    "Pending",
    "Failed"

}


# =====================================================================
# 9. PREPROCESS RAW DATA
# =====================================================================

df["Price_Numeric"] = (

    df["Price"]
    .apply(clean_price)

)


df["Quantity_Numeric"] = pd.to_numeric(

    df["Quantity"],

    errors="coerce"

)


df["Parsed_Date"] = pd.to_datetime(

    df["Transaction_Date"],

    errors="coerce"

)


product_original = (

    df["Product_Name"]
    .fillna("")
    .astype(str)
    .str.strip()

)


payment_original = (

    df["Payment_Method"]
    .fillna("")
    .astype(str)
    .str.strip()

)


status_original = (

    df["Transaction_Status"]
    .fillna("")
    .astype(str)
    .str.strip()

)


# =====================================================================
# 10. ERROR RULES
# =====================================================================


df["Error_Missing_Transaction_ID"] = (

    df["Transaction_ID"]
    .isna()
    .astype(int)

)


df["Error_Invalid_Transaction_ID"] = (

    ~df["Transaction_ID"]
    .apply(valid_transaction_id)

).astype(int)


df["Error_Duplicate_Transaction_ID"] = (

    df["Transaction_ID"]
    .duplicated(
        keep=False
    )

    &

    df["Transaction_ID"]
    .notna()

).astype(int)


df["Error_Missing_Customer_ID"] = (

    df["Customer_ID"]
    .isna()
    .astype(int)

)


df["Error_Invalid_Customer_ID"] = (

    ~df["Customer_ID"]
    .apply(valid_customer_id)

).astype(int)


df["Error_Missing_Date"] = (

    df["Transaction_Date"]
    .isna()
    .astype(int)

)


df["Error_Invalid_Date"] = (

    df["Transaction_Date"]
    .notna()

    &

    df["Parsed_Date"]
    .isna()

).astype(int)


df["Error_Missing_Product"] = (

    df["Product_Name"]
    .isna()
    .astype(int)

)


df["Error_Invalid_Product"] = (

    df["Product_Name"]
    .notna()

    &

    ~product_original.isin(
        VALID_PRODUCTS
    )

).astype(int)


df["Error_Missing_Quantity"] = (

    df["Quantity_Numeric"]
    .isna()
    .astype(int)

)


df["Error_Invalid_Quantity"] = (

    df["Quantity_Numeric"]
    .notna()

    &

    (
        df["Quantity_Numeric"]
        <= 0
    )

).astype(int)


df["Error_Missing_Price"] = (

    df["Price"]
    .isna()
    .astype(int)

)


df["Error_Invalid_Price_Format"] = (

    df["Price"]
    .notna()

    &

    df["Price_Numeric"]
    .isna()

).astype(int)


df["Error_Invalid_Price_Value"] = (

    df["Price_Numeric"]
    .notna()

    &

    (
        df["Price_Numeric"]
        <= 0
    )

).astype(int)


df["Error_Missing_Payment_Method"] = (

    df["Payment_Method"]
    .isna()
    .astype(int)

)


df["Error_Nonstandard_Payment"] = (

    df["Payment_Method"]
    .notna()

    &

    ~payment_original.isin(
        VALID_PAYMENT_METHODS
    )

).astype(int)


df["Error_Missing_Status"] = (

    df["Transaction_Status"]
    .isna()
    .astype(int)

)


df["Error_Nonstandard_Status"] = (

    df["Transaction_Status"]
    .notna()

    &

    ~status_original.isin(
        VALID_STATUSES
    )

).astype(int)


# =====================================================================
# 11. ERROR COLUMN LIST
# =====================================================================

ERROR_COLUMNS = [

    "Error_Missing_Transaction_ID",
    "Error_Invalid_Transaction_ID",
    "Error_Duplicate_Transaction_ID",

    "Error_Missing_Customer_ID",
    "Error_Invalid_Customer_ID",

    "Error_Missing_Date",
    "Error_Invalid_Date",

    "Error_Missing_Product",
    "Error_Invalid_Product",

    "Error_Missing_Quantity",
    "Error_Invalid_Quantity",

    "Error_Missing_Price",
    "Error_Invalid_Price_Format",
    "Error_Invalid_Price_Value",

    "Error_Missing_Payment_Method",
    "Error_Nonstandard_Payment",

    "Error_Missing_Status",
    "Error_Nonstandard_Status"

]


ERROR_NAMES = {

    "Error_Missing_Transaction_ID":
        "Missing Transaction ID",

    "Error_Invalid_Transaction_ID":
        "Invalid Transaction ID",

    "Error_Duplicate_Transaction_ID":
        "Duplicate Transaction ID",

    "Error_Missing_Customer_ID":
        "Missing Customer ID",

    "Error_Invalid_Customer_ID":
        "Invalid Customer ID",

    "Error_Missing_Date":
        "Missing Date",

    "Error_Invalid_Date":
        "Invalid Date",

    "Error_Missing_Product":
        "Missing Product",

    "Error_Invalid_Product":
        "Invalid Product",

    "Error_Missing_Quantity":
        "Missing Quantity",

    "Error_Invalid_Quantity":
        "Invalid Quantity",

    "Error_Missing_Price":
        "Missing Price",

    "Error_Invalid_Price_Format":
        "Invalid Price Format",

    "Error_Invalid_Price_Value":
        "Invalid Price Value",

    "Error_Missing_Payment_Method":
        "Missing Payment Method",

    "Error_Nonstandard_Payment":
        "Nonstandard Payment Method",

    "Error_Missing_Status":
        "Missing Status",

    "Error_Nonstandard_Status":
        "Nonstandard Status"

}


# =====================================================================
# 12. CREATE WRONGNESS TARGET
# =====================================================================

df["Error_Count"] = (

    df[
        ERROR_COLUMNS
    ]
    .sum(
        axis=1
    )

)


df["Is_Anomaly"] = (

    df["Error_Count"] > 0

).astype(int)


# =====================================================================
# 13. CREATE ERROR DESCRIPTION
# =====================================================================


def get_error_type(row):

    errors = []

    for column in ERROR_COLUMNS:

        if row[column] == 1:

            errors.append(
                ERROR_NAMES[column]
            )

    if len(errors) == 0:

        return "Clean"

    return " | ".join(
        errors
    )


df["Error_Type"] = df.apply(

    get_error_type,

    axis=1

)


# =====================================================================
# 14. FEATURE ENGINEERING
# =====================================================================

features = pd.DataFrame(
    index=df.index
)


# Transaction ID

features["transaction_id_missing"] = (

    df["Transaction_ID"]
    .isna()
    .astype(int)

)


features["transaction_id_length"] = (

    df["Transaction_ID"]
    .fillna("")
    .astype(str)
    .str.len()

)


features["transaction_id_valid"] = (

    df["Transaction_ID"]
    .apply(valid_transaction_id)
    .astype(int)

)


features["transaction_id_duplicate"] = (

    df["Transaction_ID"]
    .duplicated(
        keep=False
    )
    .astype(int)

)


# Customer ID

features["customer_missing"] = (

    df["Customer_ID"]
    .isna()
    .astype(int)

)


features["customer_id_length"] = (

    df["Customer_ID"]
    .fillna("")
    .astype(str)
    .str.len()

)


# Date

features["date_missing"] = (

    df["Transaction_Date"]
    .isna()
    .astype(int)

)


features["date_valid"] = (

    df["Parsed_Date"]
    .notna()
    .astype(int)

)


features["year"] = (

    df["Parsed_Date"]
    .dt.year
    .fillna(0)

)


features["month"] = (

    df["Parsed_Date"]
    .dt.month
    .fillna(0)

)


features["day"] = (

    df["Parsed_Date"]
    .dt.day
    .fillna(0)

)


# Product

features["product_missing"] = (

    df["Product_Name"]
    .isna()
    .astype(int)

)


features["product_length"] = (

    product_original
    .str.len()

)


features["product_valid"] = (

    product_original
    .isin(
        VALID_PRODUCTS
    )
    .astype(int)

)


# Quantity

features["quantity_missing"] = (

    df["Quantity_Numeric"]
    .isna()
    .astype(int)

)


features["quantity_value"] = (

    df["Quantity_Numeric"]
    .fillna(0)

)


features["quantity_positive"] = (

    df["Quantity_Numeric"]
    .gt(0)
    .fillna(False)
    .astype(int)

)


features["quantity_negative"] = (

    df["Quantity_Numeric"]
    .lt(0)
    .fillna(False)
    .astype(int)

)


# Price

features["price_missing"] = (

    df["Price"]
    .isna()
    .astype(int)

)


features["price_value"] = (

    df["Price_Numeric"]
    .fillna(0)

)


features["price_positive"] = (

    df["Price_Numeric"]
    .gt(0)
    .fillna(False)
    .astype(int)

)


features["price_negative"] = (

    df["Price_Numeric"]
    .lt(0)
    .fillna(False)
    .astype(int)

)


features["price_has_dollar"] = (

    df["Price"]
    .fillna("")
    .astype(str)
    .str.contains(
        r"\$",
        regex=True
    )
    .astype(int)

)


features["price_length"] = (

    df["Price"]
    .fillna("")
    .astype(str)
    .str.len()

)


# Payment

features["payment_missing"] = (

    df["Payment_Method"]
    .isna()
    .astype(int)

)


features["payment_valid"] = (

    payment_original
    .isin(
        VALID_PAYMENT_METHODS
    )
    .astype(int)

)


features["payment_length"] = (

    payment_original
    .str.len()

)


# Status

features["status_missing"] = (

    df["Transaction_Status"]
    .isna()
    .astype(int)

)


features["status_valid"] = (

    status_original
    .isin(
        VALID_STATUSES
    )
    .astype(int)

)


features["status_length"] = (

    status_original
    .str.len()

)


features = features.replace(

    [
        np.inf,
        -np.inf
    ],

    np.nan

)


features = features.fillna(
    0
)


features = features.astype(
    np.float32
)


X = features.copy()


y = (

    df["Is_Anomaly"]
    .astype(
        np.int32
    )

)


print(
    "\nTotal generated features:",
    X.shape[1]
)


# =====================================================================
# 15. TRAIN TEST SPLIT
# =====================================================================

X_train_full, X_test, y_train_full, y_test = train_test_split(

    X,
    y,

    test_size=0.20,

    random_state=SEED,

    stratify=y

)


# =====================================================================
# 16. AIS FEATURE SELECTION
# =====================================================================
#
# Artificial Immune System
# Clonal Selection Algorithm
#
# Each antibody:
#
# [1, 0, 1, 1, 0, ...]
#
# 1 = feature selected
# 0 = feature ignored
#
# Fitness:
# F1 score + small penalty for too many features
#
# =====================================================================


print("\n" + "=" * 80)
print("ARTIFICIAL IMMUNE SYSTEM FEATURE OPTIMIZATION")
print("=" * 80)


NUMBER_OF_FEATURES = X_train_full.shape[1]


# AIS configuration

AIS_POPULATION_SIZE = 20

AIS_GENERATIONS = 15

AIS_ELITE_COUNT = 5

AIS_CLONES_PER_ELITE = 4

AIS_MUTATION_RATE = 0.10


# =====================================================================
# 17. CREATE AIS OPTIMIZATION SUBSET
# =====================================================================
#
# We use part of training data during AIS feature optimization
# so that the optimization remains practical.
#
# =====================================================================

AIS_SAMPLE_SIZE = min(
    30000,
    len(X_train_full)
)


ais_split = StratifiedShuffleSplit(

    n_splits=1,

    train_size=AIS_SAMPLE_SIZE,

    random_state=SEED

)


ais_index, _ = next(

    ais_split.split(
        X_train_full,
        y_train_full
    )

)


X_ais = (

    X_train_full
    .iloc[
        ais_index
    ]
    .reset_index(
        drop=True
    )

)


y_ais = (

    y_train_full
    .iloc[
        ais_index
    ]
    .reset_index(
        drop=True
    )

)


X_ais_train, X_ais_val, y_ais_train, y_ais_val = train_test_split(

    X_ais,
    y_ais,

    test_size=0.25,

    random_state=SEED,

    stratify=y_ais

)


# =====================================================================
# 18. AIS FITNESS FUNCTION
# =====================================================================


def evaluate_antibody(mask):

    selected_indices = np.where(
        mask == 1
    )[0]


    # Prevent zero selected features

    if len(selected_indices) == 0:

        return 0.0


    X_tr = X_ais_train.iloc[
        :,
        selected_indices
    ]


    X_vl = X_ais_val.iloc[
        :,
        selected_indices
    ]


    scaler_temp = StandardScaler()


    X_tr_scaled = scaler_temp.fit_transform(
        X_tr
    )


    X_vl_scaled = scaler_temp.transform(
        X_vl
    )


    model_temp = LogisticRegression(

        max_iter=500,

        random_state=SEED,

        class_weight="balanced",

        solver="liblinear"

    )


    try:

        model_temp.fit(

            X_tr_scaled,

            y_ais_train

        )


        predictions = model_temp.predict(
            X_vl_scaled
        )


        score = f1_score(

            y_ais_val,

            predictions,

            zero_division=0

        )


    except:

        return 0.0


    # ---------------------------------------------
    # Feature reduction bonus
    # ---------------------------------------------

    feature_ratio = (

        len(selected_indices)
        /
        NUMBER_OF_FEATURES

    )


    reduction_bonus = (

        1.0
        -
        feature_ratio

    )


    # Fitness:
    # 95% predictive quality
    # 5% feature reduction

    fitness = (

        0.95
        *
        score

        +

        0.05
        *
        reduction_bonus

    )


    return float(
        fitness
    )


# =====================================================================
# 19. CREATE INITIAL AIS POPULATION
# =====================================================================


def create_random_antibody():

    antibody = np.random.randint(

        0,
        2,

        size=NUMBER_OF_FEATURES

    )


    if np.sum(
        antibody
    ) == 0:

        antibody[
            np.random.randint(
                0,
                NUMBER_OF_FEATURES
            )
        ] = 1


    return antibody


population = [

    create_random_antibody()

    for _ in range(
        AIS_POPULATION_SIZE
    )

]


# Include all-feature antibody

population[0] = np.ones(

    NUMBER_OF_FEATURES,

    dtype=int

)


# =====================================================================
# 20. AIS CLONAL SELECTION PROCESS
# =====================================================================


ais_history = []


best_global_mask = None

best_global_fitness = -1


for generation in range(

    AIS_GENERATIONS

):


    print(
        f"\nAIS Generation "
        f"{generation + 1}/"
        f"{AIS_GENERATIONS}"
    )


    fitness_scores = []


    for antibody in population:

        fitness = evaluate_antibody(
            antibody
        )

        fitness_scores.append(
            fitness
        )


    fitness_scores = np.array(
        fitness_scores
    )


    order = np.argsort(
        fitness_scores
    )[::-1]


    population = [

        population[index]

        for index in order

    ]


    fitness_scores = (

        fitness_scores[
            order
        ]

    )


    generation_best = float(
        fitness_scores[0]
    )


    generation_mask = population[0]


    selected_count = int(
        np.sum(
            generation_mask
        )
    )


    print(
        "Best fitness:",
        round(
            generation_best,
            6
        )
    )


    print(
        "Selected features:",
        selected_count,
        "/",
        NUMBER_OF_FEATURES
    )


    ais_history.append(

        {

            "Generation":
                generation + 1,

            "Best_Fitness":
                generation_best,

            "Selected_Features":
                selected_count

        }

    )


    if generation_best > best_global_fitness:

        best_global_fitness = (
            generation_best
        )

        best_global_mask = (

            generation_mask.copy()

        )


    # ------------------------------------------------------------
    # Select elites
    # ------------------------------------------------------------

    elites = [

        population[i].copy()

        for i in range(
            min(
                AIS_ELITE_COUNT,
                len(population)
            )
        )

    ]


    # ------------------------------------------------------------
    # Clone and mutate elites
    # ------------------------------------------------------------

    new_population = []


    # Preserve elites

    new_population.extend(
        elites
    )


    for elite_rank, elite in enumerate(
        elites
    ):


        for clone_number in range(

            AIS_CLONES_PER_ELITE

        ):


            clone = elite.copy()


            # Better antibodies mutate slightly less

            mutation_probability = (

                AIS_MUTATION_RATE
                *
                (
                    1
                    +
                    elite_rank
                    /
                    max(
                        1,
                        AIS_ELITE_COUNT
                    )
                )

            )


            for feature_index in range(

                NUMBER_OF_FEATURES

            ):


                if np.random.rand() < mutation_probability:

                    clone[
                        feature_index
                    ] = (

                        1
                        -
                        clone[
                            feature_index
                        ]

                    )


            if np.sum(
                clone
            ) == 0:

                clone[
                    np.random.randint(
                        0,
                        NUMBER_OF_FEATURES
                    )
                ] = 1


            new_population.append(
                clone
            )


    # ------------------------------------------------------------
    # Add random antibodies for immune diversity
    # ------------------------------------------------------------

    while len(
        new_population
    ) < AIS_POPULATION_SIZE:

        new_population.append(

            create_random_antibody()

        )


    population = new_population[
        :AIS_POPULATION_SIZE
    ]


# =====================================================================
# 21. AIS FINAL SELECTED FEATURES
# =====================================================================

selected_feature_indices = np.where(

    best_global_mask == 1

)[0]


selected_features = [

    X.columns[index]

    for index in selected_feature_indices

]


print("\n" + "=" * 80)
print("AIS FEATURE SELECTION COMPLETE")
print("=" * 80)


print(
    "\nBest AIS fitness:",
    best_global_fitness
)


print(
    "\nSelected feature count:",
    len(
        selected_features
    )
)


print("\nAIS selected features:")


for feature in selected_features:

    print(
        " -",
        feature
    )


# =====================================================================
# 22. SAVE AIS OPTIMIZATION HISTORY
# =====================================================================

ais_history_df = pd.DataFrame(
    ais_history
)


ais_history_df.to_csv(

    AIS_HISTORY_CSV_PATH,

    index=False

)


print(
    "\nSaved:",
    AIS_HISTORY_CSV_PATH
)


# =====================================================================
# 23. AIS OPTIMIZATION HISTORY GRAPH
# =====================================================================

plt.figure(
    figsize=(10, 6)
)


plt.plot(

    ais_history_df[
        "Generation"
    ],

    ais_history_df[
        "Best_Fitness"
    ],

    marker="o"

)


plt.xlabel(
    "AIS Generation"
)

plt.ylabel(
    "Best Fitness"
)

plt.title(
    "AIS Clonal Selection Optimization"
)

plt.grid(
    True,
    alpha=0.3
)

plt.tight_layout()


plt.savefig(

    AIS_HISTORY_GRAPH_PATH,

    dpi=300,

    bbox_inches="tight"

)


print(
    "Saved:",
    AIS_HISTORY_GRAPH_PATH
)


plt.show()

plt.close()


# =====================================================================
# 24. SAVE FEATURE SELECTION CSV
# =====================================================================

feature_selection_df = pd.DataFrame(

    {

        "Feature":
            list(
                X.columns
            ),

        "AIS_Selected":
            [

                1
                if feature in selected_features
                else 0

                for feature
                in X.columns

            ]

    }

)


feature_selection_df.to_csv(

    FEATURE_SELECTION_CSV_PATH,

    index=False

)


print(
    "Saved:",
    FEATURE_SELECTION_CSV_PATH
)


# =====================================================================
# 25. AIS FEATURE SELECTION GRAPH
# =====================================================================

plt.figure(
    figsize=(12, 8)
)


plt.barh(

    feature_selection_df[
        "Feature"
    ],

    feature_selection_df[
        "AIS_Selected"
    ]

)


plt.xlabel(
    "Selected by AIS (1 = Yes, 0 = No)"
)

plt.ylabel(
    "Feature"
)

plt.title(
    "AIS Feature Selection Result"
)

plt.tight_layout()


plt.savefig(

    FEATURE_SELECTION_GRAPH_PATH,

    dpi=300,

    bbox_inches="tight"

)


print(
    "Saved:",
    FEATURE_SELECTION_GRAPH_PATH
)


plt.show()

plt.close()


# =====================================================================
# 26. USE ONLY AIS SELECTED FEATURES
# =====================================================================

X_train = X_train_full[
    selected_features
].copy()


X_test_selected = X_test[
    selected_features
].copy()


# =====================================================================
# 27. SCALE FINAL FEATURES
# =====================================================================

scaler = StandardScaler()


X_train_scaled = scaler.fit_transform(
    X_train
)


X_test_scaled = scaler.transform(
    X_test_selected
)


# =====================================================================
# 28. CLASS WEIGHTS
# =====================================================================

class_counts = np.bincount(
    y_train_full
)


class_weight = None


if len(class_counts) == 2:

    total = len(
        y_train_full
    )


    class_weight = {

        0:

            total
            /
            (
                2
                *
                class_counts[0]
            ),

        1:

            total
            /
            (
                2
                *
                class_counts[1]
            )

    }


# =====================================================================
# 29. BUILD FINAL ANN
# =====================================================================

model = Sequential(

    [

        tf.keras.Input(

            shape=(
                X_train_scaled.shape[1],
            )

        ),

        Dense(
            128,
            activation="relu"
        ),

        BatchNormalization(),

        Dropout(
            0.30
        ),

        Dense(
            64,
            activation="relu"
        ),

        BatchNormalization(),

        Dropout(
            0.25
        ),

        Dense(
            32,
            activation="relu"
        ),

        Dropout(
            0.20
        ),

        Dense(
            16,
            activation="relu"
        ),

        Dense(
            1,
            activation="sigmoid"
        )

    ]

)


model.compile(

    optimizer=tf.keras.optimizers.Adam(
        learning_rate=0.001
    ),

    loss="binary_crossentropy",

    metrics=[

        "accuracy",

        tf.keras.metrics.Precision(
            name="precision"
        ),

        tf.keras.metrics.Recall(
            name="recall"
        ),

        tf.keras.metrics.AUC(
            name="auc"
        )

    ]

)


model.summary()


# =====================================================================
# 30. CALLBACKS
# =====================================================================

early_stopping = EarlyStopping(

    monitor="val_loss",

    patience=7,

    restore_best_weights=True,

    verbose=1

)


reduce_lr = ReduceLROnPlateau(

    monitor="val_loss",

    factor=0.5,

    patience=3,

    min_lr=1e-6,

    verbose=1

)


# =====================================================================
# 31. TRAIN FINAL AIS-ANN MODEL
# =====================================================================

print("\n" + "=" * 80)
print("TRAINING FINAL AIS-OPTIMIZED ANN")
print("=" * 80)


history = model.fit(

    X_train_scaled,
    y_train_full,

    validation_split=0.20,

    epochs=50,

    batch_size=256,

    callbacks=[

        early_stopping,
        reduce_lr

    ],

    class_weight=class_weight,

    verbose=1

)


# =====================================================================
# 32. PREDICTIONS
# =====================================================================

prediction_probability = (

    model.predict(

        X_test_scaled,

        verbose=0

    )

    .flatten()

)


prediction_label = (

    prediction_probability
    >= 0.50

).astype(int)


# =====================================================================
# 33. METRICS
# =====================================================================

accuracy = accuracy_score(

    y_test,
    prediction_label

)


precision = precision_score(

    y_test,
    prediction_label,

    zero_division=0

)


recall = recall_score(

    y_test,
    prediction_label,

    zero_division=0

)


f1 = f1_score(

    y_test,
    prediction_label,

    zero_division=0

)


try:

    roc_auc = roc_auc_score(

        y_test,
        prediction_probability

    )

except:

    roc_auc = 0.0


cm = confusion_matrix(

    y_test,
    prediction_label

)


print("\n" + "=" * 80)
print("AIS MODEL RESULTS")
print("=" * 80)


print(
    f"\nAccuracy : {accuracy * 100:.2f}%"
)

print(
    f"Precision: {precision * 100:.2f}%"
)

print(
    f"Recall   : {recall * 100:.2f}%"
)

print(
    f"F1 Score : {f1 * 100:.2f}%"
)

print(
    f"ROC-AUC  : {roc_auc * 100:.2f}%"
)


print("\nClassification Report:")


print(

    classification_report(

        y_test,
        prediction_label,

        target_names=[
            "Clean",
            "Anomaly"
        ],

        zero_division=0

    )

)


# =====================================================================
# 34. AIS ACCURACY GRAPH
# =====================================================================

epochs_range = range(

    1,

    len(
        history.history[
            "accuracy"
        ]
    ) + 1

)


plt.figure(
    figsize=(10, 6)
)


plt.plot(

    epochs_range,

    history.history[
        "accuracy"
    ],

    marker="o",

    markersize=3,

    label="Training Accuracy"

)


plt.plot(

    epochs_range,

    history.history[
        "val_accuracy"
    ],

    marker="o",

    markersize=3,

    label="Validation Accuracy"

)


plt.xlabel(
    "Epoch"
)

plt.ylabel(
    "Accuracy"
)

plt.title(
    "AIS-Optimized ANN Accuracy"
)

plt.legend()

plt.grid(
    True,
    alpha=0.3
)

plt.tight_layout()


plt.savefig(

    ACCURACY_GRAPH_PATH,

    dpi=300,

    bbox_inches="tight"

)


print(
    "Saved:",
    ACCURACY_GRAPH_PATH
)


plt.show()

plt.close()


# =====================================================================
# 35. AIS CONFUSION MATRIX HEATMAP
# =====================================================================

plt.figure(
    figsize=(7, 6)
)


plt.imshow(
    cm
)


plt.title(
    "AIS Confusion Matrix Heatmap"
)


plt.xlabel(
    "Predicted Class"
)

plt.ylabel(
    "Actual Class"
)


plt.xticks(

    [0, 1],

    [
        "Clean",
        "Anomaly"
    ]

)


plt.yticks(

    [0, 1],

    [
        "Clean",
        "Anomaly"
    ]

)


for i in range(
    cm.shape[0]
):

    for j in range(
        cm.shape[1]
    ):

        plt.text(

            j,
            i,

            str(
                cm[i, j]
            ),

            ha="center",
            va="center",

            fontsize=14

        )


plt.colorbar()

plt.tight_layout()


plt.savefig(

    HEATMAP_PATH,

    dpi=300,

    bbox_inches="tight"

)


print(
    "Saved:",
    HEATMAP_PATH
)


plt.show()

plt.close()


# =====================================================================
# 36. AIS COMPARISON GRAPH
# =====================================================================

metric_names = [

    "Accuracy",
    "Precision",
    "Recall",
    "F1 Score",
    "ROC-AUC"

]


metric_values = [

    accuracy * 100,
    precision * 100,
    recall * 100,
    f1 * 100,
    roc_auc * 100

]


plt.figure(
    figsize=(10, 6)
)


bars = plt.bar(

    metric_names,

    metric_values

)


plt.ylabel(
    "Score (%)"
)

plt.xlabel(
    "Metric"
)

plt.title(
    "AIS Model Performance Comparison"
)


plt.ylim(
    0,
    105
)


for bar, value in zip(

    bars,
    metric_values

):

    plt.text(

        bar.get_x()
        +
        bar.get_width() / 2,

        value + 1,

        f"{value:.2f}%",

        ha="center"

    )


plt.grid(
    axis="y",
    alpha=0.3
)

plt.tight_layout()


plt.savefig(

    COMPARISON_GRAPH_PATH,

    dpi=300,

    bbox_inches="tight"

)


print(
    "Saved:",
    COMPARISON_GRAPH_PATH
)


plt.show()

plt.close()


# =====================================================================
# 37. AIS RESULT CSV
# =====================================================================

result_records = []


for column in ERROR_COLUMNS:

    error_count = int(

        df[column]
        .sum()

    )


    percentage = (

        error_count
        /
        len(df)
        *
        100

    )


    result_records.append(

        {

            "Error_Type":
                ERROR_NAMES[column],

            "Error_Count":
                error_count,

            "Percentage":
                round(
                    percentage,
                    4
                )

        }

    )


result_df = pd.DataFrame(
    result_records
)


result_df = result_df.sort_values(

    "Error_Count",

    ascending=False

).reset_index(
    drop=True
)


result_df.to_csv(

    RESULT_CSV_PATH,

    index=False

)


print(
    "Saved:",
    RESULT_CSV_PATH
)


# =====================================================================
# 38. AIS RESULT GRAPH
# =====================================================================

plt.figure(
    figsize=(13, 8)
)


bars = plt.barh(

    result_df[
        "Error_Type"
    ],

    result_df[
        "Error_Count"
    ]

)


plt.xlabel(
    "Number of Transactions"
)

plt.ylabel(
    "Error Type"
)

plt.title(
    "AIS Financial Data Quality Error Distribution"
)


plt.gca().invert_yaxis()


for bar in bars:

    value = bar.get_width()

    plt.text(

        value,

        bar.get_y()
        +
        bar.get_height() / 2,

        f" {int(value)}",

        va="center"

    )


plt.grid(
    axis="x",
    alpha=0.3
)

plt.tight_layout()


plt.savefig(

    RESULT_GRAPH_PATH,

    dpi=300,

    bbox_inches="tight"

)


print(
    "Saved:",
    RESULT_GRAPH_PATH
)


plt.show()

plt.close()


# =====================================================================
# 39. AIS PREDICTION CSV
# =====================================================================

prediction_df = (

    original_df
    .loc[
        X_test.index
    ]
    .copy()

)


prediction_df[
    "Actual_Label"
] = y_test.values


prediction_df[
    "Actual_Result"
] = np.where(

    y_test.values == 1,

    "Anomaly",

    "Clean"

)


prediction_df[
    "AIS_Prediction_Probability"
] = prediction_probability


prediction_df[
    "AIS_Predicted_Label"
] = prediction_label


prediction_df[
    "AIS_Predicted_Result"
] = np.where(

    prediction_label == 1,

    "Anomaly",

    "Clean"

)


prediction_df[
    "Correct_Prediction"
] = (

    y_test.values
    ==
    prediction_label

)


prediction_df[
    "Error_Count"
] = (

    df.loc[
        X_test.index,
        "Error_Count"
    ].values

)


prediction_df[
    "Error_Type"
] = (

    df.loc[
        X_test.index,
        "Error_Type"
    ].values

)


prediction_df.to_csv(

    PREDICTION_CSV_PATH,

    index=False

)


print(
    "Saved:",
    PREDICTION_CSV_PATH
)


# =====================================================================
# 40. AIS PREDICTION GRAPH
# =====================================================================

number_to_plot = min(

    100,

    len(
        prediction_df
    )

)


plot_df = prediction_df.iloc[
    :number_to_plot
]


x_positions = np.arange(
    number_to_plot
)


plt.figure(
    figsize=(15, 7)
)


plt.plot(

    x_positions,

    plot_df[
        "Actual_Label"
    ],

    marker="o",

    markersize=4,

    label="Actual Label"

)


plt.plot(

    x_positions,

    plot_df[
        "AIS_Prediction_Probability"
    ],

    marker=".",

    markersize=4,

    label="AIS Prediction Probability"

)


plt.axhline(

    y=0.5,

    linestyle="--",

    label="Decision Threshold"

)


plt.xlabel(
    "Test Transaction"
)

plt.ylabel(
    "Prediction Value"
)

plt.title(
    "AIS Actual vs Predicted Wrongness"
)


plt.yticks(

    [
        0,
        0.5,
        1
    ],

    [
        "Clean",
        "Threshold",
        "Anomaly"
    ]

)


plt.legend()

plt.grid(
    True,
    alpha=0.3
)

plt.tight_layout()


plt.savefig(

    PREDICTION_GRAPH_PATH,

    dpi=300,

    bbox_inches="tight"

)


print(
    "Saved:",
    PREDICTION_GRAPH_PATH
)


plt.show()

plt.close()


# =====================================================================
# 41. SAVE LABELED DATASET
# =====================================================================

labeled_df = original_df.copy()


for column in ERROR_COLUMNS:

    labeled_df[
        column
    ] = df[
        column
    ]


labeled_df[
    "Error_Count"
] = df[
    "Error_Count"
]


labeled_df[
    "Error_Type"
] = df[
    "Error_Type"
]


labeled_df[
    "Is_Anomaly"
] = df[
    "Is_Anomaly"
]


labeled_df.to_csv(

    LABELED_CSV_PATH,

    index=False

)


print(
    "Saved:",
    LABELED_CSV_PATH
)


# =====================================================================
# 42. SAVE AIS H5 MODEL
# =====================================================================

model.save(
    H5_PATH
)


print(
    "Saved:",
    H5_PATH
)


# =====================================================================
# 43. SAVE AIS PKL
# =====================================================================

pipeline_data = {

    "algorithm":

        "Artificial Immune System "
        "Clonal Selection + ANN",

    "scaler":

        scaler,

    "all_features":

        list(
            X.columns
        ),

    "selected_features":

        selected_features,

    "ais_best_mask":

        best_global_mask.tolist(),

    "ais_best_fitness":

        float(
            best_global_fitness
        ),

    "prediction_threshold":

        0.50,

    "valid_products":

        list(
            VALID_PRODUCTS
        ),

    "valid_payment_methods":

        list(
            VALID_PAYMENT_METHODS
        ),

    "valid_statuses":

        list(
            VALID_STATUSES
        )

}


joblib.dump(

    pipeline_data,

    PKL_PATH

)


print(
    "Saved:",
    PKL_PATH
)


# =====================================================================
# 44. SAVE AIS YAML
# =====================================================================

yaml_config = {

    "project": {

        "name":
            "AIS Financial Transaction Data Quality Prediction",

        "algorithm":
            "Artificial Immune System + ANN",

        "dataset":
            str(
                CSV_PATH
            )

    },


    "ais": {

        "algorithm":
            "Clonal Selection Algorithm",

        "population_size":
            AIS_POPULATION_SIZE,

        "generations":
            AIS_GENERATIONS,

        "elite_count":
            AIS_ELITE_COUNT,

        "clones_per_elite":
            AIS_CLONES_PER_ELITE,

        "mutation_rate":
            AIS_MUTATION_RATE,

        "best_fitness":
            float(
                best_global_fitness
            ),

        "original_feature_count":
            int(
                NUMBER_OF_FEATURES
            ),

        "selected_feature_count":
            int(
                len(
                    selected_features
                )
            ),

        "selected_features":
            selected_features

    },


    "ann": {

        "hidden_layers": [

            128,
            64,
            32,
            16

        ],

        "optimizer":
            "Adam",

        "loss":
            "binary_crossentropy",

        "batch_size":
            256,

        "maximum_epochs":
            50

    },


    "metrics": {

        "accuracy":
            float(
                accuracy
            ),

        "precision":
            float(
                precision
            ),

        "recall":
            float(
                recall
            ),

        "f1_score":
            float(
                f1
            ),

        "roc_auc":
            float(
                roc_auc
            )

    }

}


with open(

    YAML_PATH,

    "w",

    encoding="utf-8"

) as file:

    yaml.safe_dump(

        yaml_config,

        file,

        sort_keys=False,

        allow_unicode=True

    )


print(
    "Saved:",
    YAML_PATH
)


# =====================================================================
# 45. SAVE AIS JSON
# =====================================================================

history_json = {

    key: [

        float(
            value
        )

        for value in values

    ]

    for key, values
    in history.history.items()

}


json_data = {

    "project":

        "AIS Financial Transaction Data Quality Prediction",

    "algorithm":

        "Artificial Immune System Clonal Selection + ANN",

    "dataset_records":

        int(
            len(df)
        ),

    "original_features":

        int(
            NUMBER_OF_FEATURES
        ),

    "selected_features_count":

        int(
            len(
                selected_features
            )
        ),

    "selected_features":

        selected_features,

    "ais_best_fitness":

        float(
            best_global_fitness
        ),

    "metrics": {

        "accuracy":
            float(
                accuracy
            ),

        "precision":
            float(
                precision
            ),

        "recall":
            float(
                recall
            ),

        "f1_score":
            float(
                f1
            ),

        "roc_auc":
            float(
                roc_auc
            )

    },

    "confusion_matrix":

        cm.tolist(),

    "ais_history":

        ais_history,

    "training_history":

        history_json

}


with open(

    JSON_PATH,

    "w",

    encoding="utf-8"

) as file:

    json.dump(

        json_data,

        file,

        indent=4,

        ensure_ascii=False

    )


print(
    "Saved:",
    JSON_PATH
)


# =====================================================================
# 46. VERIFY SAVED MODEL
# =====================================================================

loaded_model = tf.keras.models.load_model(
    H5_PATH
)


verification = (

    loaded_model.predict(

        X_test_scaled[:5],

        verbose=0

    )

    .flatten()

)


print("\nVerification predictions:")


for i, probability in enumerate(

    verification,

    start=1

):

    result = (

        "ANOMALY"

        if probability >= 0.5

        else "CLEAN"

    )


    print(

        f"{i}. "
        f"{probability:.4f} "
        f"-> {result}"

    )


# =====================================================================
# 47. FINAL FILE CHECK
# =====================================================================

generated_files = [

    H5_PATH,
    PKL_PATH,
    YAML_PATH,
    JSON_PATH,

    ACCURACY_GRAPH_PATH,
    HEATMAP_PATH,
    COMPARISON_GRAPH_PATH,

    RESULT_CSV_PATH,
    RESULT_GRAPH_PATH,

    PREDICTION_CSV_PATH,
    PREDICTION_GRAPH_PATH,

    FEATURE_SELECTION_CSV_PATH,
    FEATURE_SELECTION_GRAPH_PATH,

    AIS_HISTORY_CSV_PATH,
    AIS_HISTORY_GRAPH_PATH,

    LABELED_CSV_PATH

]


print("\n" + "=" * 80)
print("AIS PROJECT COMPLETED")
print("=" * 80)


for number, file_path in enumerate(

    generated_files,

    start=1

):


    status = (

        "SUCCESS"

        if file_path.exists()

        else "FAILED"

    )


    print(

        f"{number:02d}. "
        f"{file_path.name:45s} "
        f"[{status}]"

    )


# =====================================================================
# 48. FINAL SUMMARY
# =====================================================================

print("\n" + "=" * 80)
print("FINAL AIS RESULTS")
print("=" * 80)


print(
    f"\nAIS best fitness : "
    f"{best_global_fitness:.4f}"
)


print(
    f"Original features: "
    f"{NUMBER_OF_FEATURES}"
)


print(
    f"Selected features: "
    f"{len(selected_features)}"
)


print(
    f"\nAccuracy : "
    f"{accuracy * 100:.2f}%"
)


print(
    f"Precision: "
    f"{precision * 100:.2f}%"
)


print(
    f"Recall   : "
    f"{recall * 100:.2f}%"
)


print(
    f"F1 Score : "
    f"{f1 * 100:.2f}%"
)


print(
    f"ROC-AUC  : "
    f"{roc_auc * 100:.2f}%"
)


print(
    "\nOutput directory:"
)


print(
    PROJECT_DIR
)


print(
    "\nAll files use the ais_ prefix."
)


print("=" * 80)
```
